## İki boyutlu histogramlar ve binleme


````python
# bivariate_normal_sample.py
mean = [0, 0]
cov = [[1, 1], [1, 2]]
x, y = rng.multivariate_normal(mean, cov, 10000).T

````

### plt.hist2d: İki boyutlu histogram

İki boyutlu histogram çizmenin doğrudan yolu Matplotlib'in plt.hist2d işlevidir (aşağıdaki şekle bakın):


````python
# hist2d_plot.py
plt.hist2d(x, y, bins=30)
cb = plt.colorbar()
cb.set_label('counts in bin')

````

plt.hist gibi plt.hist2d'nin de grafik ve binlemeyi ince ayarlayan seçenekleri vardır; docstring'te güzel özetlenmiştir.
    plt.hist'in np.histogram karşılığı olduğu gibi plt.hist2d'nin karşılığı np.histogram2d'dir:


````python
# np_histogram2d.py
counts, xedges, yedges = np.histogram2d(x, y, bins=30)
print(counts.shape)

````

İki boyuttan fazla boyutta binlemenin genellemesi için np.histogramdd işlevine bakın.

### plt.hexbin: Altıgen binleme

İki boyutlu histogram eksenler boyunca karelerden oluşan bir döşeme oluşturur.
    Bu tür döşeme için doğal bir başka şekil düzenli altıgendir.
    Matplotlib bu amaçla plt.hexbin sunar; iki boyutlu veriyi altıgen ızgarada gruplar (aşağıdaki şekle bakın):


````python
# hexbin_plot.py
plt.hexbin(x, y, gridsize=30)
cb = plt.colorbar(label='count in bin')

````

plt.hexbin'in ek seçenekleri vardır: her nokta için ağırlık belirtme, her kutudaki çıktıyı herhangi bir NumPy toplamına (ağırlık ortalaması, ağırlık standart sapması vb.) çevirme.

### Çekirdek yoğunluk tahmini

Çok boyutlu yoğunlukları tahmin etmenin ve göstermenin bir başka yaygın yolu çekirdek yoğunluk tahmini (KDE)'dir.
    Bu konu kitapta ayrıntılı ele alınır; şimdilik KDE'nin uzaydaki noktaları «yayarak» sonucu toplayıp pürüzsüz bir fonksiyon elde etme yolu olarak düşünülebilir.
    scipy.stats paketinde çok hızlı ve basit bir KDE uygulaması vardır.
    KDE kullanımına kısa bir örnek (aşağıdaki şekle bakın):


````python
# kde_plot.py
from scipy.stats import gaussian_kde

# fit an array of size [Ndim, Nsamples]
data = np.vstack([x, y])
kde = gaussian_kde(data)

# evaluate on a regular grid
xgrid = np.linspace(-3.5, 3.5, 40)
ygrid = np.linspace(-6, 6, 40)
Xgrid, Ygrid = np.meshgrid(xgrid, ygrid)
Z = kde.evaluate(np.vstack([Xgrid.ravel(), Ygrid.ravel()]))

# Plot the result as an image
plt.imshow(Z.reshape(Xgrid.shape),
           origin='lower', aspect='auto',
           extent=[-3.5, 3.5, -6, 6])
cb = plt.colorbar()
cb.set_label("density")

````

KDE'nin ayrıntı ile pürüzsüzlük arasında kaydıran bir yumuşatma uzunluğu vardır (yaygın önyargı–varyans ödünleşiminin bir örneği).
    Uygun yumuşatma uzunluğu seçimi geniş bir literatüre sahiptir; gaussian_kde girdi verisi için neredeyse optimal bir uzunluk bulmaya çalışır.

SciPy ekosisteminde farklı güçlü ve zayıf yönleri olan başka KDE uygulamaları da vardır; örneğin sklearn.neighbors.KernelDensity ve statsmodels.nonparametric.KDEMultivariate.

KDE tabanlı görselleştirmelerde Matplotlib genelde fazla ayrıntılı olabilir.
    Seaborn ile Görselleştirme bölümünde tartışılan Seaborn kütüphanesi KDE tabanlı görselleştirmeler için çok daha kompakt bir API sunar.
