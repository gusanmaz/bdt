Orijinal: 04.06 Customizing Legends

Lejantlar bir görselleştirmeye anlam katar; çeşitli çizim öğelerine etiket atar.
    Basit bir lejant oluşturmayı daha önce gördük; burada Matplotlib'de lejantın konumunu ve görünümünü özelleştirmeye bakacağız.

En basit lejant plt.legend komutuyla oluşturulur; bu komut etiketlenmiş tüm çizim öğeleri için otomatik lejant üretir (aşağıdaki şekle bakın):


```python
# import_plt_style.py
import matplotlib.pyplot as plt
plt.style.use('seaborn-whitegrid')

```

```python
# matplotlib_inline.py
%matplotlib inline
import numpy as np

```

```python
# legend_basic.py
x = np.linspace(0, 10, 1000)
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), '-b', label='Sine')
ax.plot(x, np.cos(x), '--r', label='Cosine')
ax.axis('equal')
leg = ax.legend()

```

> **Not**
>

Böyle bir lejantı birçok yönden özelleştirebiliriz.
    Örneğin konumu belirtebilir ve çerçeveyi açabiliriz (aşağıdaki şekle bakın):


```python
# legend_loc_frame.py
ax.legend(loc='upper left', frameon=True)
fig

```

Lejanttaki sütun sayısını ncol ile belirleyebiliriz (aşağıdaki şekil):


```python
# legend_ncol.py
ax.legend(loc='lower center', ncol=2)
fig

```

## Lejant için öğe seçimi

Varsayılan olarak lejant, çizimdeki tüm etiketli öğeleri içerir.
    İstenen bu değilse, plot komutlarının döndürdüğü nesnelerle hangi öğe ve etiketlerin görüneceğini ince ayar yapabiliriz.
    plt.plot birden fazla çizgiyi aynı anda oluşturabilir ve oluşan Line2D örneklerinin listesini döndürür.
    Bunlardan herhangi birini plt.legend'a vermek, hangilerinin gösterileceğini ve hangi etiketlerin kullanılacağını belirtir (aşağıdaki şekil):


```python
# legend_lines_list.py
y = np.sin(x[:, np.newaxis] + np.pi * np.arange(0, 2, 0.5))
lines = plt.plot(x, y)

# lines is a list of plt.Line2D instances
plt.legend(lines[:2], ['first', 'second'], frameon=True);

```

Pratikte genelde ilk yöntem daha net: lejantta göstermek istediğiniz öğelere doğrudan label verin (aşağıdaki şekil):


```python
# legend_labels.py
plt.plot(x, y[:, 0], label='first')
plt.plot(x, y[:, 1], label='second')
plt.plot(x, y[:, 2:])
plt.legend(frameon=True);

```

Lejant, label özniteliği ayarlanmamış tüm öğeleri yok sayar.

## Nokta boyutu için lejant

Bazen varsayılan lejant veri görselleştirmesi için yeterli olmaz.
    Örneğin nokta boyutunu kullanarak verinin belirli özelliklerini işaretliyorsanız, bunu yansıtan bir lejant isteyebilirsiniz.
    Burada Kaliforniya şehirlerinin nüfusunu nokta boyutuyla gösteriyoruz.
    Nokta boyutlarının ölçeğini gösteren bir lejant oluşturmak için etiketli ama veri içermeyen çizimler kullanacağız (aşağıdaki şekil):


```python
# Uncomment to download the data
# url = ('https://raw.githubusercontent.com/jakevdp/PythonDataScienceHandbook/'
#        'master/notebooks/data/california_cities.csv')
# !cd data && curl -O {url}

```

```python
# cities_scatter_legend.py
import pandas as pd
cities = pd.read_csv('data/california_cities.csv')

# Extract the data we're interested in
lat, lon = cities['latd'], cities['longd']
population, area = cities['population_total'], cities['area_total_km2']

# Scatter the points, using size and color but no label
plt.scatter(lon, lat, label=None,
            c=np.log10(population), cmap='viridis',
            s=area, linewidth=0, alpha=0.5)
plt.axis('equal')
plt.xlabel('longitude')
plt.ylabel('latitude')
plt.colorbar(label='log$_{10}$(population)')
plt.clim(3, 7)

# Here we create a legend:
# we'll plot empty lists with the desired size and label
for area in [100, 300, 500]:
    plt.scatter([], [], c='k', alpha=0.3, s=area,
                label=str(area) + ' km$^2$')
plt.legend(scatterpoints=1, frameon=False, labelspacing=1, title='City Area')

plt.title('California Cities: Area and Population');

```

### 🧪 Şimdi deneyin

      Farklı boyutlarda sahte noktalarla özel bir lejant oluşturun.
          
      for s, lab in [(20, 'küçük'), (80, 'büyük')]:
    plt.scatter([], [], s=s, c='gray', alpha=0.5, label=lab)
plt.legend(scatterpoints=1, title='Boyut')

Lejant her zaman çizimdeki bir nesneye referans verir; belirli bir şekli göstermek istiyorsanız onu çizmeniz gerekir.
    Burada istediğimiz nesneler (gri daireler) çizimde olmadığı için boş listeler çizerek sahte öğeler oluşturuyoruz.
    Lejant yalnızca label belirtilmiş öğeleri listeler.

Boş listeler çizerek lejantın alacağı etiketli nesneler yaratılır; lejant böylece yararlı bilgi verir.
    Bu strateji daha gelişmiş görselleştirmeler için kullanışlı olabilir.

## Birden fazla lejant

Bazen aynı eksene birden fazla lejant eklemek istersiniz.
    Maalesef Matplotlib bunu kolay yapmaz: standart legend arayüzüyle yalnızca tek bir lejant oluşturulabilir.
    İkinci bir plt.legend veya ax.legend çağrısı birincisinin üzerine yazar.
    Bunu, sıfırdan yeni bir lejant sanatçısı (Artist, Matplotlib'in görsel öğe taban sınıfı) oluşturup düşük seviyeli ax.add_artist ile ikinci sanatçıyı elle ekleyerek aşabiliriz (aşağıdaki şekil):


```python
# multiple_legends.py
fig, ax = plt.subplots()

lines = []
styles = ['-', '--', '-.', ':']
x = np.linspace(0, 10, 1000)

for i in range(4):
    lines += ax.plot(x, np.sin(x - i * np.pi / 2),
                     styles[i], color='black')
ax.axis('equal')

# Specify the lines and labels of the first legend
ax.legend(lines[:2], ['line A', 'line B'], loc='upper right')

# Create the second legend and add the artist manually
from matplotlib.legend import Legend
leg = Legend(ax, lines[2:], ['line C', 'line D'], loc='lower right')
ax.add_artist(leg);

```

Bu, her Matplotlib çizimini oluşturan düşük seviyeli sanatçı nesnelerine kısa bir bakıştır.
    ax.legend kaynak koduna bakarsanız (Jupyter'de ax.legend?? ile yapabilirsiniz) işlevin uygun bir Legend sanatçısı oluşturup legend_ özniteliğine kaydettiğini ve çizim çizilirken figüre eklendiğini görürsünüz.
