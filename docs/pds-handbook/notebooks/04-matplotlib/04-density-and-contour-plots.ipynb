## Üç boyutlu bir fonksiyonu görselleştirme

İlk örneğimiz $z = f(x, y)$ fonksiyonu için kontur grafiği gösterir; fonksiyon şöyle seçilmiştir:


````python
# contour_function_f.py
def f(x, y):
    return np.sin(x) ** 10 + np.cos(10 + y * x) * np.cos(x)

````

Kontur grafiği plt.contour ile oluşturulur. Üç argüman alır: x değerleri ızgarası, y değerleri ızgarası ve z değerleri ızgarası.


````python
# meshgrid_xyz.py
x = np.linspace(0, 5, 50)
y = np.linspace(0, 5, 40)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)

````

> **Not**
>

Şimdi yalnızca çizgili standart kontur grafiğine bakalım (aşağıdaki şekle bakın):


````python
# contour_black.py
plt.contour(X, Y, Z, colors='black');

````

Tek renk kullanıldığında negatif değerler kesikli, pozitif değerler düz çizgiyle gösterilir.
    Alternatif olarak cmap argümanıyla çizgiler renklendirilebilir.
    Burada veri aralığında 20 eşit aralıklı çizgi istiyoruz (aşağıdaki şekle bakın):


````python
# contour_colormap.py
plt.contour(X, Y, Z, 20, cmap='RdGy');

````

RdGy (Red–Gray, kırmızı–gri) renk haritasını seçtik; sıfır etrafında pozitif ve negatif sapma olan
    ıraksak (divergent) veri için iyi bir seçimdir.
    Matplotlib'te birçok renk haritası vardır; IPython'da plt.cm modülünde sekme tamamlama ile kolayca gezinebilirsiniz:


```
plt.cm.<TAB>
```


Grafiğimiz daha iyi görünüyor; çizgiler arası boşluklar dikkat dağıtıcı olabilir.
    Bunu plt.contourf ile dolu kontur grafiğine geçerek değiştirebiliriz; sözdizimi büyük ölçüde plt.contour ile aynıdır.

Ek olarak renk bilgisi için etiketli ek bir eksen oluşturan plt.colorbar komutunu ekleyeceğiz (aşağıdaki şekle bakın):


````python
# contourf_colorbar.py
plt.contourf(X, Y, Z, 20, cmap='RdGy')
plt.colorbar();

````

### 🧪 Şimdi deneyin

🧪 Kontur + renk çubuğu
      Aynı f(x,y) fonksiyonu için contourf ve farklı bir cmap (ör. 'viridis') deneyin:
          
      import matplotlib.pyplot as plt
import numpy as np

def f(x, y):
    return np.sin(x) ** 10 + np.cos(10 + y * x) * np.cos(x)

x = np.linspace(0, 5, 50)
y = np.linspace(0, 5, 40)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

plt.contourf(X, Y, Z, 25, cmap='viridis')
plt.colorbar(label='z')
plt.title('Dolu kontur — viridis')

Renk çubuğu siyah bölgelerin «tepe», kırmızı bölgelerin «vadi» olduğunu netleştirir.

Bu grafikte renk adımları ayrık olduğu için lekeli görünebilir; her zaman istenen bu değildir.
    Çok yüksek kontur sayısı ayarlanabilir ama verimsiz olur: Matplotlib her seviye için yeni bir çokgen çizer.
    Daha pürüzsüz gösterim için plt.imshow ve interpolation argümanı kullanılabilir (aşağıdaki şekle bakın):


````python
# imshow_interpolation.py
plt.imshow(Z, extent=[0, 5, 0, 5], origin='lower', cmap='RdGy',
           interpolation='gaussian', aspect='equal')
plt.colorbar();

````

plt.imshow ile birkaç olası tuzak vardır:

Son olarak kontur ve görüntü grafiklerini birleştirmek bazen yararlıdır.
    Örneğin kısmen saydam arka plan görüntüsü (alpha ile) ve üzerine plt.clabel ile etiketlenmiş konturlar kullanalım (aşağıdaki şekle bakın):


````python
# contour_imshow_combo.py
contours = plt.contour(X, Y, Z, 3, colors='black')
plt.clabel(contours, inline=True, fontsize=8)

plt.imshow(Z, extent=[0, 5, 0, 5], origin='lower',
           cmap='RdGy', alpha=0.5)
plt.colorbar();

````

plt.contour, plt.contourf ve plt.imshow birleşimi,
    bu tür üç boyutlu veriyi iki boyutlu grafikte göstermek için neredeyse sınırsız olanak sunar.
