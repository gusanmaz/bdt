## plt.plot ile saçılım grafikleri

Önceki bölümde çizgi grafikleri için plt.plot/ax.plot kullanmıştık.
    Aynı işlev saçılım grafikleri de üretebilir (aşağıdaki şekle bakın):


````python
# scatter_plot_o.py
x = np.linspace(0, 10, 30)
y = np.sin(x)

plt.plot(x, y, 'o', color='black');

````

Fonksiyon çağrısındaki üçüncü argüman, çizimde kullanılan sembol türünü temsil eden bir karakterdir.
    Çizgi stilini '-' veya '--' ile belirttiğiniz gibi, işaretçi stilinin de kısa dize kodları vardır.
    Kullanılabilir sembollerin tam listesi plt.plot dokümantasyonunda veya Matplotlib'in
    çevrimiçi dokümantasyonunda görülebilir.
    Çoğu sezgiseldir; daha yaygın olanların bir kısmı burada gösterilmiştir (aşağıdaki şekle bakın):


````python
# marker_styles_demo.py
rng = np.random.default_rng(0)
for marker in ['o', '.', ',', 'x', '+', 'v', '^', '<', '>', 's', 'd']:
    plt.plot(rng.random(2), rng.random(2), marker, color='black',
             label="marker='{0}'".format(marker))
plt.legend(numpoints=1, fontsize=13)
plt.xlim(0, 1.8);

````

Daha fazla seçenek için bu karakter kodları, noktaları birleştiren çizgiyle birlikte kullanılabilir (aşağıdaki şekle bakın):


````python
# scatter_line_marker.py
plt.plot(x, y, '-ok');

````

plt.plot'a verilen ek anahtar sözcük argümanları çizgi ve işaretçilerin geniş özellik aralığını belirler (aşağıdaki şekle bakın):


````python
# scatter_styled_markers.py
plt.plot(x, y, '-p', color='gray',
         markersize=15, linewidth=4,
         markerfacecolor='white',
         markeredgecolor='gray',
         markeredgewidth=2)
plt.ylim(-1.2, 1.2);

````

Bu tür seçenekler plt.plot'u Matplotlib'de iki boyutlu grafikler için temel araç yapar.
    Kullanılabilir seçeneklerin tam açıklaması için
    plt.plot dokümantasyonuna bakın.

## plt.scatter ile saçılım grafikleri

Saçılım grafiği oluşturmanın ikinci, daha güçlü yolu plt.scatter işlevidir;
    plt.plot'a çok benzer kullanılır (aşağıdaki şekle bakın):


````python
# scatter_basic.py
plt.scatter(x, y, marker='o');

````

plt.scatter'ın plt.plot'tan temel farkı, her noktanın özelliklerinin
    (boyut, yüz rengi, kenar rengi vb.) ayrı ayrı kontrol edilebilmesi veya veriye eşlenebilmesidir.

Çok renkli ve boyutlu rastgele bir saçılım grafiği oluşturalım.
    Üst üste binen sonuçları daha iyi görmek için alpha anahtar sözcüğüyle saydamlığı ayarlayacağız (aşağıdaki şekle bakın):


````python
# scatter_color_size.py
rng = np.random.default_rng(0)
x = rng.normal(size=100)
y = rng.normal(size=100)
colors = rng.random(100)
sizes = 1000 * rng.random(100)

plt.scatter(x, y, c=colors, s=sizes, alpha=0.3)
plt.colorbar();  # show color scale

````

> **Not**
>

color argümanının otomatik olarak bir renk ölçeğine eşlendiğine dikkat edin
    (burada colorbar komutuyla gösterilir); s (size) argümanı piksel cinsindendir.
    Böylece renk ve nokta boyutu görselleştirmede bilgi taşıyabilir; çok boyutlu veriyi görselleştirmek için kullanılır.

Örneğin Scikit-Learn'deki Iris veri setini kullanabiliriz: her örnek, taç yaprak ve çanak yaprak boyutları dikkatle ölçülmüş üç çiçek türünden biridir (aşağıdaki şekle bakın):


````python
# iris_scatter.py
from sklearn.datasets import load_iris
iris = load_iris()
features = iris.data.T

plt.scatter(features[0], features[1], alpha=0.4,
            s=100*features[3], c=iris.target, cmap='viridis')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1]);

````

### 🧪 Şimdi deneyin

🧪 Renk ve boyutla saçılım deneyin
      Rastgele 50 nokta üretin; rengi ve boyutu ayrı değişkenlerle eşleyip colorbar ekleyin:
          
      import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)
x = rng.normal(size=50)
y = rng.normal(size=50)
c = rng.random(50)
s = 500 * rng.random(50)

plt.scatter(x, y, c=c, s=s, alpha=0.5, cmap='viridis')
plt.colorbar()
plt.xlabel('x')
plt.ylabel('y')

Bu grafiğin tam renkli sürümü kitabın çevrimiçi sürümünde mevcuttur.

Bu saçılım grafiği verinin dört farklı boyutunu aynı anda keşfetmemizi sağladı:
    her noktanın (x, y) konumu çanak yaprak uzunluğu ve genişliğine karşılık gelir;
    nokta boyutu taç yaprak genişliğiyle, renk ise çiçek türüyle ilişkilidir.
    Bu tür çok renkli ve çok özellikli saçılım grafikleri hem keşif hem sunum için yararlıdır.

## plot ve scatter: verimlilik notu

plt.plot ve plt.scatter'daki farklı özellikler dışında neden birini diğerine tercih edersiniz?
    Küçük veri kümelerinde pek fark etmez; ancak birkaç bin noktadan büyük veri kümelerinde plt.plot,
    plt.scatter'dan belirgin şekilde daha verimli olabilir.
    Nedeni: plt.scatter her nokta için farklı boyut ve/veya renk çizebilir; oluşturucu her noktayı ayrı ayrı inşa etmek zorundadır.
    plt.plot'ta ise her noktanın işaretçileri aynıdır; görünüm yalnızca bir kez belirlenir.
    Büyük veri kümelerinde bu fark performansı büyük ölçüde etkileyebilir; bu yüzden büyük veri kümelerinde plt.plot tercih edilmelidir.
