Orijinal: 04.08 Multiple Subplots

Bazen verinin farklı görünümlerini yan yana karşılaştırmak yararlıdır.
    Matplotlib bunu tek bir figür içinde birden fazla alt grafik (subplot) düzenleyerek sağlar.


```python
# matplotlib_inline.py
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use('seaborn-white')
import numpy as np

```

## plt.axes: Elle alt grafikler

Eksen oluşturmanın en temel yolu plt.axes kullanmaktır.
    Varsayılan olarak bu, tüm figürü kaplayan standart bir eksen oluşturur; farklı bir konum ve boyut için plt.axes'e bir dizi verilebilir (aşağıdaki şekil):


```python
# axes_manual.py
ax1 = plt.axes()  # standard axes
ax2 = plt.axes([0.65, 0.65, 0.2, 0.2])

```

Nesne yönelimli arayüzdeki karşılığı fig.add_axes'tir.
    İki dikey istiflenmiş eksen oluşturmak için bunu kullanalım (aşağıdaki şekil):


```python
# add_axes_stacked.py
fig = plt.figure()
ax1 = fig.add_axes([0.1, 0.5, 0.8, 0.4],
                   xticklabels=[], ylim=(-1.2, 1.2))
ax2 = fig.add_axes([0.1, 0.1, 0.8, 0.4],
                   ylim=(-1.2, 1.2))

x = np.linspace(0, 10)
ax1.plot(np.sin(x))
ax2.plot(np.cos(x));

```

Artık iki eksenimiz var (üstteki işaret etiketleri yok); üst panelin altı (konum 0.5) alt panelin üstüyle (0.1 + 0.4) örtüşüyor.

## plt.subplot: Basit alt grafik ızgaraları

Hizalı sütun veya satır alt grafikler yaygındır; Matplotlib bunları kolaylaştıran birkaç yardımcı rutin sunar.
    En düşük seviye plt.subplot'tur; ızgarada tek bir alt grafik oluşturur.
    Komut üç tamsayı alır — satır sayısı, sütun sayısı ve bu şemada (sol üstten sağ alta) oluşturulacak grafiğin indeksi (aşağıdaki şekil):


```python
# subplot_loop.py
for i in range(1, 7):
    plt.subplot(2, 3, i)
    plt.text(0.5, 0.5, str((2, 3, i)),
             fontsize=18, ha='center')

```

### 🧪 Şimdi deneyin

🧪 Alt grafik ızgarası
      2×2 ızgara oluşturup her hücreye metin yazın.
          
      fig, axes = plt.subplots(2, 2)
for ax in axes.flat:
    ax.text(0.5, 0.5, 'panel', ha='center', fontsize=14)

Alt grafikler arasındaki boşluğu ayarlamak için plt.subplots_adjust kullanılabilir.
    Aşağıdaki kod nesne yönelimli eşdeğeri fig.add_subplot'u kullanır (aşağıdaki şekil):


```python
# subplots_adjust.py
fig = plt.figure()
fig.subplots_adjust(hspace=0.4, wspace=0.4)
for i in range(1, 7):
    ax = fig.add_subplot(2, 3, i)
    ax.text(0.5, 0.5, str((2, 3, i)),
           fontsize=18, ha='center')

```

Burada plt.subplots_adjust'in hspace ve wspace bağımsız değişkenlerini kullandık; bunlar alt grafik boyutunun birimleri cinsinden dikey ve yatay boşluğu belirtir (burada boşluk alt grafik genişlik ve yüksekliğinin %40'ı).

## plt.subplots: Tüm ızgarayı tek seferde

Büyük bir alt grafik ızgarası oluştururken ve iç panellerde eksen etiketlerini gizlemek istediğinizde yukarıdaki yöntem çabuk yorucu olur.
    Bu iş için plt.subplots daha kolaydır (sonundaki s'e dikkat). Tek alt grafik yerine tek satırda tam bir ızgara oluşturur ve NumPy dizisi olarak döndürür.
    Bağımsız değişkenler satır ve sütun sayısıdır; isteğe bağlı sharex ve sharey eksenler arası ölçek ilişkisini belirler.

Aynı satırdaki tüm eksenler y ölçeğini, aynı sütundakiler x ölçeğini paylaşan $2 \times 3$ bir ızgara oluşturalım (aşağıdaki şekil):


```python
# subplots_share.py
fig, ax = plt.subplots(2, 3, sharex='col', sharey='row')

```

> **Not**
>

sharex ve sharey ile ızgaranın iç etiketleri otomatik kaldırıldı; daha temiz bir çizim elde edildi.
    Dönen eksen dizisi, standart dizi indeksleme ile istenen ekseni seçmeyi kolaylaştırır (aşağıdaki şekil):


```python
# axes are in a two-dimensional array, indexed by [row, col]
for i in range(2):
    for j in range(3):
        ax[i, j].text(0.5, 0.5, str((i, j)),
                      fontsize=18, ha='center')
fig

```

plt.subplot'a kıyasla plt.subplots Python'un sıfırdan başlayan indekslemesiyle daha tutarlıdır; plt.subplot MATLAB tarzı birden başlayan indeks kullanır.

## plt.GridSpec: Daha karmaşık düzenler

Birden fazla satır ve sütunu kapsayan alt grafiklere geçmek için en iyi araç plt.GridSpec'tir.
    plt.GridSpec kendi başına çizim oluşturmaz; plt.subplot tarafından tanınan bir arayüzdür.
    Örneğin belirli genişlik ve yükseklik aralıklı iki satır üç sütunlu bir GridSpec şöyle görünür:


```python
# gridspec_empty.py


```

```python
# gridspec_create.py
grid = plt.GridSpec(2, 3, wspace=0.4, hspace=0.3)

```

Bundan sonra alt grafik konum ve boyutlarını tanıdık Python dilimleme sözdizimiyle belirtebiliriz (aşağıdaki şekil):


```python
# gridspec_slice.py
plt.subplot(grid[0, 0])
plt.subplot(grid[0, 1:])
plt.subplot(grid[1, :2])
plt.subplot(grid[1, 2]);

```

Bu esnek hizalama geniş kullanım alanına sahiptir; en çok aşağıdaki şekildeki gibi çok eksenli histogram düzenleri için kullanırım:


```python
# Create some normally distributed data
mean = [0, 0]
cov = [[1, 1], [1, 2]]
rng = np.random.default_rng(1701)
x, y = rng.multivariate_normal(mean, cov, 3000).T

# Set up the axes with GridSpec
fig = plt.figure(figsize=(6, 6))
grid = plt.GridSpec(4, 4, hspace=0.2, wspace=0.2)
main_ax = fig.add_subplot(grid[:-1, 1:])
y_hist = fig.add_subplot(grid[:-1, 0], xticklabels=[], sharey=main_ax)
x_hist = fig.add_subplot(grid[-1, 1:], yticklabels=[], sharex=main_ax)

# Scatter points on the main axes
main_ax.plot(x, y, 'ok', markersize=3, alpha=0.2)

# Histogram on the attached axes
x_hist.hist(x, 40, histtype='stepfilled',
            orientation='vertical', color='gray')
x_hist.invert_yaxis()

y_hist.hist(y, 40, histtype='stepfilled',
            orientation='horizontal', color='gray')
y_hist.invert_xaxis()

```
