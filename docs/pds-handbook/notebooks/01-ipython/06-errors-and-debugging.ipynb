Orijinal: Errors and Debugging

Kod geliştirme ve veri analizi her zaman bir miktar deneme yanılma gerektirir; IPython bu süreci kolaylaştıran araçlar içerir. Bu bölüm kısaca Python'un istisna raporlamasını kontrol etme seçeneklerini, ardından kodda hataları ayıklama araçlarını ele alır.

## İstisnaları kontrol etmek: %xmode

Python betiği çoğu zaman başarısız olduğunda bir istisna fırlatır. Yorumlayıcı bu istisnalardan birine çarptığında hatanın nedeni hakkında bilgi traceback (izleme) içinde bulunur; bu bilgiye Python içinden erişilebilir. %xmode magic fonksiyonu ile IPython, istisna yükseltildiğinde yazdırılan bilgi miktarını kontrol etmenizi sağlar. Aşağıdaki kodu düşünün:


````python
# func_ornek.py
def func1(a, b):
    return a / b

def func2(x):
    a = x
    b = x - 1
    return func1(a, b)

````

````python
# func2_cagri.py
func2(1)

````

func2 çağrısı bir hataya yol açar; yazdırılan izlemeyi okuyarak tam olarak ne olduğunu görebilirsiniz. Varsayılan modda bu izleme, hataya giden her adımın bağlamını gösteren birkaç satır içerir. %xmode magic fonksiyonu (exception mode, istisna modu kısaltması) ile yazdırılan bilgiyi değiştirebiliriz.

%xmode tek bir argüman alır: mod. Üç seçenek vardır: Plain, Context ve Verbose. Varsayılan Context'tir; az önce gösterilen çıktıya benzer sonuç verir. Plain daha özdür ve daha az bilgi verir:


```python
# xmode_plain.ipynb
%xmode Plain
```

*(Jupyter/IPython veya özel ortam gerekir — web sayfasında salt okunur.)*


````python
# func2_plain.py
func2(1)

````

> **Not**
>

Verbose modu, çağrılan fonksiyonların argümanları dahil ek bilgi ekler:


```python
# xmode_verbose.ipynb
%xmode Verbose
```

*(Jupyter/IPython veya özel ortam gerekir — web sayfasında salt okunur.)*


````python
# func2_verbose.py
func2(1)

````

Bu ek bilgi istisnanın neden yükseltildiğini daraltmaya yardımcı olur. O halde neden her zaman Verbose kullanmayalım? Kod karmaşıklaştıkça bu tür traceback'ler son derece uzun olabilir. Bağlama göre bazen Plain veya Context modunun kısalığı daha kullanışlıdır.

## Hata ayıklama: Traceback okumak yetmediğinde

Etkileşimli hata ayıklama için standart Python aracı pdb (Python debugger)'dır. Bu debugger kullanıcının kodda satır satır ilerlemesine ve daha zor bir hatanın nedenini görmesine olanak tanır. IPython ile geliştirilmiş sürümü ipdb'dir.

Her iki debugger'ı başlatmanın ve kullanmanın birçok yolu vardır; burada tam olarak ele almayacağız. Daha fazla bilgi için bu iki aracın çevrimiçi dokümantasyonuna bakın.

IPython'da belki de hata ayıklamanın en uygun arayüzü %debug magic komutudur. Bir istisnadan sonra çağırırsanız otomatik olarak istisnanın oluştuğu noktada etkileşimli bir hata ayıklama istemi açar. ipdb istemi yığının mevcut durumunu keşfetmenize, kullanılabilir değişkenlere bakmanıza ve hatta Python komutları çalıştırmanıza izin verir!

En son istisnaya bakalım ve temel işlemler yapalım. a ve b değerlerini yazdırıp ardından hata ayıklama oturumundan çıkmak için quit yazın:


```python
# debug.ipynb
%debug
```

*(Jupyter/IPython veya özel ortam gerekir — web sayfasında salt okunur.)*


Etkileşimli debugger bundan çok daha fazlasını sağlar — yığın içinde yukarı ve aşağı adımlayıp değişken değerlerini orada da inceleyebilirsiniz:


```python
%debug
```

*(Jupyter/IPython veya özel ortam gerekir — web sayfasında salt okunur.)*


Böylece yalnızca hataya neyin yol açtığını değil, hataya hangi fonksiyon çağrılarının öncülük ettiğini de hızla bulabiliriz.

İstisna yükseltildiğinde debugger'ın otomatik açılmasını istiyorsanız %pdb magic fonksiyonu ile bu davranışı açabilirsiniz:


```python
# pdb_on.ipynb
%xmode Plain
%pdb on
func2(1)
```

*(Jupyter/IPython veya özel ortam gerekir — web sayfasında salt okunur.)*


Son olarak, baştan etkileşimli modda çalıştırmak istediğiniz bir betiğiniz varsa %run -d komutuyla çalıştırıp next ile kod satırlarında adım adım ilerleyebilirsiniz.
