Orijinal: Keyboard Shortcuts in the IPython Shell

Bilgisayarda zaman geçiriyorsanız iş akışınızda klavye kısayollarına muhtemelen yer vermişsinizdir. En tanıdıkları Cmd-c / Cmd-v (veya Ctrl-c / Ctrl-v) kopyala-yapıştırdır. Güç kullanıcıları daha da ileri gider: Emacs, Vim gibi editörler karmaşık tuş kombinasyonlarıyla çok geniş işlem yelpazesi sunar.

IPython kabuğu bu kadar ileri gitmez; komut yazarken hızlı gezinme için bir dizi klavye kısayolu sunar. Bunların bir kısmı tarayıcı tabanlı defterlerde de çalışır; bu bölüm özellikle IPython kabuğundaki kısayollara odaklanır.

Alıştığınızda ellerinizi «ev» klavye konumundan kaldırmadan belirli komutları hızla çalıştırmak çok yararlı olur. Emacs veya Linux tarzı kabuk deneyiminiz varsa aşağıdakiler tanıdık gelecektir. Kısayolları birkaç kategoride toplayacağız: gezinme, metin girişi, komut geçmişi ve çeşitli.

## Metin Girişi Kısayolları

Backspace önceki karakteri silmek için tanıdıktır; tuşa uzanmak küçük bir parmak jimnastiği ister ve tek seferde bir karakter siler. IPython'da yazdığınız metnin bir bölümünü silmek için birkaç kısayol vardır; en hemen işe yarayanları tüm satırı kesmektir. Ctrl-b ve Ctrl-d'yi Backspace yerine kullanmaya başladığınızda ikinci doğaya dönüşmüştür!

## Komut Geçmişi Kısayolları

Buradaki en etkili kısayollardan biri IPython'un komut geçmişinde gezinmesidir. Geçmiş yalnızca mevcut oturumla sınırlı değildir: tüm komut geçmişiniz IPython profil dizininizde bir SQLite veritabanında saklanır. Önceki komutlara en doğrudan yol yukarı/aşağı ok tuşlarıdır; başka seçenekler de vardır:

Geriye arama özellikle yararlıdır. Daha önce square adlı bir fonksiyon tanımlamıştık. Yeni bir IPython kabuğunda Python geçmişinde geriye arayıp bu tanımı yeniden bulalım. IPython terminalinde Ctrl-r'ye bastığınızda şu istemi görürsünüz:


```
In [1]:
(reverse-i-search)`':
```


Bu istemde karakter yazmaya başladığınızda IPython, eşleşen en son komutu otomatik doldurur:


```
In [1]:
(reverse-i-search)`sqa': square??
```


İstediğiniz zaman daha fazla karakter ekleyerek aramayı daraltabilir veya Ctrl-r'ye tekrar basarak bir sonraki eşleşmeyi arayabilirsiniz. Daha önce takip ettiyseniz Ctrl-r'ye iki kez daha basınca:


```
In [1]:
(reverse-i-search)`sqa': def square(a):
    """Return the square of a"""
    return a ** 2
```


Aradığınız komutu bulunca Return'a basın; arama biter ve komutu oturumda kullanmaya devam edebilirsiniz:


```
In [1]: def square(a):
    """Return the square of a"""
    return a ** 2

In [2]: square(2)
Out[2]: 4
```


Ctrl-p/Ctrl-n veya yukarı/aşağı ok ile de benzer şekilde geçmişte arayabilirsiniz; ancak yalnızca satır başındaki karakterlerle eşleşir. Yani def yazıp Ctrl-p'ye basarsanız geçmişte def ile başlayan en son komut bulunur.
