IPython/Jupyter'ın en yararlı işlevlerinden biri, kullanıcı ile işini verimli yapmasına yardımcı olacak dokümantasyon ve arama arasındaki mesafeyi kısaltmaktır. Karmaşık sorularda web araması hâlâ rol oynar; şaşırtıcı miktarda bilgi yalnızca IPython ile bulunabilir. IPython'un birkaç tuş vuruşuyla yanıtlayabildiği soru örnekleri:

Burada IPython kabuğu ve Jupyter Notebook'un bu bilgiye hızlı erişim için sağladığı araçları ele alacağız: dokümantasyonu keşfetmek için ?, kaynak kodu için ?? ve otomatik tamamlama için Tab tuşu.

### Dokümantasyona ? ile Erişim

Python dili ve veri bilimi ekosistemi kullanıcı düşünülerek kurulmuştur; bunun büyük parçası dokümantasyona erişimdir. Her Python nesnesi, çoğu zaman nesnenin özünü ve kullanımını özetleyen bir docstring (belge dizisi) içeren bir string referansı taşır. Python'un yerleşik help fonksiyonu bu bilgiye erişir ve yazdırır. Örneğin yerleşik len fonksiyonunun dokümantasyonu:


````python
# help_len.py
help(len)

````

```
Help on built-in function len in module builtins:

len(obj, /)
    Return the number of items in a container.
```


Yorumlayıcınıza bağlı olarak bu bilgi satır içi metin veya ayrı bir pencerede görünebilir.

Bir nesne hakkında yardım bulmak çok yaygın ve yararlı olduğundan IPython ve Jupyter, bu dokümantasyona ve ilgili bilgilere kısayol olarak ? karakterini sunar:


```
In [2]: len?
Signature: len(obj, /)
Docstring: Return the number of items in a container.
Type:      builtin_function_or_method
```


> **Not**
>

Bu gösterim hemen hemen her şey için geçerlidir; nesne yöntemleri dahil:


```
In [3]: L = [1, 2, 3]
In [4]: L.insert?
Signature: L.insert(index, object, /)
Docstring: Insert object before index.
Type:      builtin_function_or_method
```


hatta nesnelerin kendisi için, türlerinden gelen dokümantasyonla:


```
In [5]: L?
Type:        list
String form: [1, 2, 3]
Length:      3
Docstring:
Built-in mutable sequence.

If no argument is given, the constructor creates a new empty list.
The argument must be an iterable if specified.
```


Önemlisi, kendi oluşturduğunuz fonksiyonlar ve nesneler için de çalışır! Docstring'li küçük bir fonksiyon tanımlayalım:


```
In [6]: def square(a):
  ....:     """Return the square of a."""
  ....:     return a ** 2
  ....:
```


Pyodide'da aynı tanım:


````python
# square_tanim.py
def square(a):
    """Return the square of a."""
    return a ** 2

````

Docstring oluşturmak için fonksiyonun ilk satırına bir string literal koyduk. Docstring'ler genelde çok satırlı olduğundan, gelenek olarak Python'un çok satırlı string (üç tırnak) gösterimini kullandık.

Şimdi docstring'i bulmak için ? kullanalım:


```
In [7]: square?
Signature: square(a)
Docstring: Return the square of a.
File:      <ipython-input-6>
Type:      function
```


Docstring'ler üzerinden bu hızlı dokümantasyon erişimi, yazdığınız koda her zaman satır içi belge eklemeniz için güçlü bir nedendir!

### Kaynak Koduna ?? ile Erişim

Python okunması kolay bir dil olduğundan, merak ettiğiniz nesnenin kaynak kodunu okuyarak genelde bir katman daha derin içgörü kazanırsınız. IPython ve Jupyter, çift soru işareti (??) ile kaynak koda kısayol sağlar:


```
In [8]: square??
Signature: square(a)
Source:
def square(a):
    """Return the square of a."""
    return a ** 2
File:      <ipython-input-6>
Type:      function
```


Bu kadar basit fonksiyonlarda ?? arka plandaki ayrıntıları hızla gösterir.

Bunu çok denerseniz bazen ?? kaynak göstermez: nesne Python'da değil C veya başka derlenmiş bir dilde uygulanmıştır. O zaman ?? çıktısı ? ile aynı olur. Özellikle birçok yerleşik nesne ve tipte (daha önceki len gibi) görürsünüz:


```
In [9]: len??
Signature: len(obj, /)
Docstring: Return the number of items in a container.
Type:      builtin_function_or_method
```


? ve/veya ?? kullanmak, herhangi bir Python fonksiyonu veya modülün ne yaptığı hakkında bilgi bulmanın güçlü ve hızlı yoludur.

### 🧪 Şimdi deneyin

🧪 Şimdi deneyin — Kaynak kodu inceleme
      Pyodide'da import inspect; print(inspect.getsource(square)) ile IPython'daki square?? benzeri çıktı alın.

print(inspect.getsource(square))

### Modülleri Tab Tamamlama ile Keşfetme

Bir diğer yararlı arayüz, otomatik tamamlama ve nesne, modül ve ad alanı içeriklerini keşfetmek için Tab tuşudur. Aşağıdaki örneklerde Tab'a basılması gerektiğinde <TAB> yazacağız.

#### Nesne içeriğinde tab tamamlama

Her Python nesnesinin çeşitli öznitelik ve yöntemleri vardır. Daha önceki help gibi, Python'un bunları listeleyen yerleşik dir fonksiyonu vardır; pratikte tab tamamlama arayüzü çok daha kolaydır. Bir nesnenin tüm özniteliklerini görmek için nesne adından sonra nokta (.) ve Tab:


```
In [10]: L.<TAB>
            append() count    insert   reverse
            clear    extend   pop      sort
            copy     index    remove
```


Listeyi daraltmak için adın ilk bir veya birkaç karakterini yazıp Tab'a basın:


```
In [10]: L.c<TAB>
             clear() count()
             copy()

In [10]: L.co<TAB>
              copy()  count()
```


Tek seçenek varsa Tab satırı tamamlar; örneğin L.cou<TAB> anında L.count olur.

Python'da kamusal/dış ve özel/iç öznitelik ayrımı katı değildir; gelenek olarak baştaki alt çizgi sonrakini gösterir. Netlik için özel yöntemler varsayılan listede yoktur; alt çizgiyi yazarak listeleyebilirsiniz:


```
In [10]: L._<TAB>
           __add__             __delattr__     __eq__
           __class__           __delitem__     __format__()
           __class_getitem__() __dir__()       __ge__
           __contains__        __doc__         __getattribute__
           ...
```


Kısalık için yalnızca ilk sütunlar gösterildi. Bunların çoğu Python'un özel çift alt çizgili («dunder») yöntemleridir.

#### İçe aktarırken tab tamamlama

Tab tamamlama paketlerden nesne içe aktarırken de yararlıdır. itertools paketinde co ile başlayan tüm olası içe aktarmalar:


```
In [10]: from itertools import co<TAB>
         combinations()                  compress()
         combinations_with_replacement() count()
```


Benzer şekilde sistemde hangi içe aktarmaların göründüğünü görmek için (Python oturumunuza bağlı olarak değişir):


```
In [10]: import <TAB>
            abc                 anyio
            activate_this       appdirs
            aifc                appnope
            antigravity         argon2
            ...

In [10]: import h<TAB>
            hashlib html
            heapq   http
            hmac
```


#### Tab ötesi: Joker eşleştirme

Tab tamamlama, aradığınız adın ilk birkaç karakterini biliyorsanız iyidir; ortada veya sonda eşleştirmek için az yardımcıdır. Bunun için IPython ve Jupyter, * ile joker eşleştirme sunar.

Örneğin adı Warning ile biten her nesneyi listelemek:


```
In [10]: *Warning?
BytesWarning                  RuntimeWarning
DeprecationWarning            SyntaxWarning
FutureWarning                 UnicodeWarning
ImportWarning                 UserWarning
PendingDeprecationWarning     Warning
ResourceWarning
```


* boş string dahil herhangi bir string ile eşleşir.

Adında bir yerde find geçen string yöntemini arıyorsak:


```
In [11]: str.*find*?
str.find
str.rfind
```


Bu esnek joker arama, yeni bir paketi öğrenirken veya tanıdık bir pakete yeniden alışırken belirli bir komutu bulmak için çok işime yarar.
