Orijinal: Jupyter: Beyond Normal Python

Python için birçok geliştirme ortamı seçeneği vardır; kendi işimde hangisini kullandığım sık sorulur. Cevabım bazen insanları şaşırtır: tercih ettiğim ortam IPython artı bir metin düzenleyicidir (bende ruh halime göre Emacs veya VSCode).

Jupyter, 2001'de Fernando Perez tarafından geliştirilmiş gelişmiş bir Python yorumlayıcısı olan IPython kabuğu olarak başladı; o zamandan beri Perez'in deyişiyle “araştırma bilişiminin tüm yaşam döngüsü için araçlar” sunmayı hedefleyen bir projeye dönüştü. Python veri bilimi görevimizin motoruysa, Jupyter'i etkileşimli kontrol paneli olarak düşünebilirsiniz.

Jupyter yalnızca Python'a kullanışlı bir etkileşimli arayüz değildir; dile bir dizi yararlı sözdizimi eklentisi de getirir — bu bölümde en yararlı olanları ele alacağız. Jupyter projesinin sağladığı en tanıdık arayüz muhtemelen Jupyter Notebook'tur: tarayıcı tabanlı, geliştirme, iş birliği, paylaşım ve hatta veri bilimi sonuçlarının yayımlanması için uygun bir ortamdır. Defter formatının yararlılığına örnek olarak şu an okuduğunuz sayfadan öteye bakmanıza gerek yok: bu kitabın tüm el yazması bir dizi Jupyter defteri olarak oluşturulmuştur.

Kitabın bu kısmı, veri bilimi uygulamasında yararlı olan Jupyter ve IPython özelliklerini adım adım işleyerek başlayacak; özellikle Python'un standart özelliklerinin ötesinde sundukları sözdizimine odaklanacağız. Ardından, veri bilimi kodu yazma ve kullanmada yaygın görevleri hızlandıran daha yararlı magic komutların bir kısmına biraz daha derinlemesine gireceğiz. Son olarak, veriyi anlamayı ve sonuçları paylaşmayı kolaylaştıran defter özelliklerine değineceğiz.
