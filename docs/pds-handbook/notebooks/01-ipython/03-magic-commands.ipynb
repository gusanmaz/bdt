Orijinal: IPython Magic Commands

Önceki bölüm IPython'un Python'u verimli ve etkileşimli kullanmanızı ve keşfetmenizi gösterdi. Burada IPython'un normal Python sözdiziminin üzerine eklediği geliştirmelerden bazılarını ele alıyoruz. Bunlara IPython'da magic komutlar denir ve % ile başlar. Standart veri analizindeki çeşitli yaygın sorunları özlü çözmek için tasarlanmışlardır. İki tür vardır: tek satırda çalışan satır magicleri (tek %) ve birden çok satırı işleyen hücre magicleri (çift %%). Burada kısa örnekler vereceğiz; daha odaklı tartışma ileride gelecek.

## Dış Kod Çalıştırma: %run

Daha kapsamlı kod yazmaya başladığınızda muhtemelen etkileşimli keşif için IPython, yeniden kullanmak istediğiniz kodu saklamak için bir metin düzenleyici kullanırsınız. Bu kodu yeni bir pencerede çalıştırmak yerine IPython oturumunuz içinde çalıştırmak uygun olabilir. Bunu %run magic komutu yapar.

Örneğin şu içeriğe sahip bir myscript.py dosyası oluşturduğunuzu düşünün:


````python
# file: myscript.py

def square(x):
    """square a number"""
    return x ** 2

for N in range(1, 4):
    print(f"{N} squared is {square(N)}")

````

IPython oturumunuzdan şöyle çalıştırabilirsiniz:


```
In [6]: %run myscript.py
1 squared is 1
2 squared is 4
3 squared is 9
```


> **Not**
>

Script'i çalıştırdıktan sonra içinde tanımlanan fonksiyonlar IPython oturumunuzda kullanılabilir:


```
In [7]: square(5)
Out[7]: 25
```


Kodun nasıl çalıştırılacağını ince ayar için birkaç seçenek vardır; IPython yorumlayıcısında %run? yazarak normal yolla dokümantasyona bakın.

## Kod Çalıştırma Süresi: %timeit

Yararlı bir diğer magic %timeit'tir; arkasındaki tek satırlık Python ifadesinin çalışma süresini otomatik ölçer. Örneğin liste üretecinin performansını kontrol edebiliriz:


```
In [8]: %timeit L = [n ** 2 for n in range(1000)]
430 µs ± 3.21 µs per loop (mean ± std. dev. of 7 runs, 1000 loops each)
```


%timeit'in yararı, kısa komutlarda daha sağlam sonuç için otomatik olarak birden çok çalıştırma yapmasıdır. Çok satırlı ifadeler için ikinci bir % işareti bunu çok satırlı girdiyi işleyen hücre magic'e çevirir. Örneğin eşdeğer for döngüsü:


```
In [9]: %%timeit
   ...: L = []
   ...: for n in range(1000):
   ...:     L.append(n ** 2)
   ...:
484 µs ± 5.67 µs per loop (mean ± std. dev. of 7 runs, 1000 loops each)
```


Bu örnekte liste üreteci, eşdeğer for döngüsünden yaklaşık %10 daha hızlıdır. %timeit ve diğer zamanlama/profil yöntemlerini 1.7 Profil ve Zamanlama bölümünde inceleyeceğiz.

## Magic Fonksiyonlarda Yardım: ?, %magic ve %lsmagic

Normal Python fonksiyonları gibi IPython magic fonksiyonlarının da docstring'leri vardır; bu dokümantasyona standart yolla erişilir. Örneğin %timeit magic'inin dokümantasyonu için:


```
In [10]: %timeit?
```


Diğer fonksiyonların dokümantasyonuna benzer şekilde erişilir. Mevcut magic fonksiyonların genel açıklaması ve örnekler için:


```
In [11]: %magic
```


Tüm magic fonksiyonların hızlı listesi için:


```
In [12]: %lsmagic
```
