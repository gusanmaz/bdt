Orijinal: IPython and Shell Commands

Standart Python yorumlayıcısıyla etkileşimli çalışırken can sıkıcı bir durum, Python araçlarına ve sistem komut satırı araçlarına erişmek için birden fazla pencere arasında geçiş yapmak zorunda kalmaktır. IPython bu boşluğu kapatır ve IPython terminalinden doğrudan kabuk (shell) komutlarını çalıştırmak için bir sözdizimi sunar. Sihir ünlem işaretiyle olur: bir satırda ! sonrasında görünen her şey Python çekirdeği tarafından değil, sistem komut satırı tarafından yürütülür.

Aşağıdaki tartışma Linux veya macOS gibi Unix benzeri bir sistemde olduğunuzu varsayar. Bazı örnekler Windows'ta varsayılan olarak farklı bir kabuk kullandığı için başarısız olur; ancak Windows Subsystem for Linux kullanıyorsanız buradaki örnekler doğru çalışmalıdır. Kabuk komutlarına aşina değilseniz, Software Carpentry Foundation'ın hazırladığı mükemmel Unix shell eğitimine göz atmanızı öneririm.

## Kabuğa hızlı giriş

Kabuğu/terminali/komut satırını kullanmaya dair tam bir giriş bu bölümün kapsamının çok ötesindedir; yine de yeni başlayanlar için kısa bir giriş sunuyorum. Kabuk, bilgisayarınızla metin tabanlı etkileşim kurmanın bir yoludur. 1980'lerin ortasından beri Microsoft ve Apple yaygın grafik işletim sistemlerinin ilk sürümlerini tanıttığından beri çoğu kullanıcı işletim sistemleriyle menü seçimleri ve sürükle-bırak hareketleriyle etkileşir. Ancak işletim sistemleri bu grafik arayüzlerden çok önce de vardı ve esas olarak metin girdisi dizileriyle kontrol ediliyordu: istemde kullanıcı bir komut yazardı, bilgisayar da kullanıcının söylediğini yapardı. Bu erken istem sistemleri, bugün çoğu veri bilimcinin hâlâ kullandığı kabukların ve terminallerin öncülleriydi.

Kabuğa aşina olmayan biri, birçok sonucun simge ve menülere tıklayarak da elde edilebildiği durumda neden uğraşılacağını sorabilir. Kabuk kullanıcısı başka bir soruyla yanıt verebilir: yazarak çok daha kolay başarabileceğiniz şeyler için neden simge ve menü öğelerini arayasınız? Tipik bir teknoloji tercihi çekişmesi gibi görünse de temel görevlerin ötesine geçildiğinde kabuğun gelişmiş işlerde çok daha fazla kontrol sunduğu hızla anlaşılır — öğrenme eğrisinin korkutucu olabileceğini kabul etmek gerekir.

Örnek olarak, bir kullanıcının sisteminde dizin ve dosyaları keşfettiği, oluşturduğu ve değiştirdiği bir Linux/macOS kabuk oturumu (osx:~ $ istemdir; $ sonrası yazılan komuttur; # ile başlayan metin yalnızca açıklamadır, yazmanız gerekmez):


```bash
osx:~ $ echo "hello world"             # echo, Python'daki print gibi
hello world

osx:~ $ pwd                            # pwd = çalışma dizinini yazdır
/home/jake                             # Şu an bulunduğumuz "yol"

osx:~ $ ls                             # ls = dizin içeriğini listele
notebooks  projects

osx:~ $ cd projects/                   # cd = dizin değiştir

osx:projects $ pwd
/home/jake/projects

osx:projects $ ls
datasci_book   mpld3   myproject.txt

osx:projects $ mkdir myproject          # mkdir = yeni dizin oluştur

osx:projects $ cd myproject/

osx:myproject $ mv ../myproject.txt ./  # mv = dosya taşı. myproject.txt bir üst
                                        # dizinden (../) geçerli dizine (./) taşınıyor.
osx:myproject $ ls
myproject.txt
```

*(Kabuk komutu — terminal veya Jupyter `!` ile.)*


Bunların hepsi, tanıdık işlemleri (dizin yapısında gezinme, dizin oluşturma, dosya taşıma vb.) simge ve menü yerine komut yazarak yapmanın özlü bir yoludur. Birkaç komutla (pwd, ls, cd, mkdir, cp) en yaygın dosya işlemlerinin çoğunu yapabilirsiniz; kabuk yaklaşımı bu temellerin ötesinde gerçekten güçlü hale gelir.

## IPython'da kabuk komutları

Herhangi bir standart kabuk komutu IPython'da başına ! karakteri eklenerek doğrudan kullanılabilir. Örneğin ls, pwd ve echo komutları şöyle çalıştırılır:


```
In [1]: !ls
myproject.txt

In [2]: !pwd
/home/jake/projects/myproject

In [3]: !echo "printing from the shell"
printing from the shell
```


## Kabuğa ve kabuktan değer aktarmak

Kabuk komutları yalnızca IPython'dan çağrılmakla kalmaz; IPython ad alanıyla da etkileşime girebilir. Örneğin herhangi bir kabuk komutunun çıktısını atama operatörü = ile bir Python listesine kaydedebilirsiniz:


```
In [4]: contents = !ls

In [5]: print(contents)
['myproject.txt']

In [6]: directory = !pwd

In [7]: print(directory)
['/Users/jakevdp/notebooks/tmp/myproject']
```


Bu sonuçlar liste olarak değil, IPython'da tanımlı özel bir kabuk dönüş tipi olarak gelir:


```
In [8]: type(directory)
IPython.utils.text.SList
```


Bu tip Python listesine çok benzer görünür ve davranır; ancak grep ve fields metodları ile sonuçları aramak, süzmek ve uygun biçimde göstermek için s, n ve p özellikleri gibi ek işlevlere sahiptir. Ayrıntılar için IPython'un yerleşik yardım özelliklerini kullanabilirsiniz.

Diğer yönde — Python değişkenlerini kabuğa geçirmek — {varname} sözdizimiyle mümkündür:


```
In [9]: message = "hello from Python"

In [10]: !echo {message}
hello from Python
```


Süslü parantezler değişken adını içerir; kabuk komutunda değişkenin içeriğiyle değiştirilir.

## Kabukla ilgili magic komutlar

IPython kabuk komutlarıyla bir süre oynadıysanız dosya sisteminde gezinmek için !cd kullanamadığınızı fark edebilirsiniz:


```
In [11]: !pwd
/home/jake/projects/myproject

In [12]: !cd ..

In [13]: !pwd
/home/jake/projects/myproject
```


Nedeni, notebook'taki kabuk komutlarının komuttan komuta durumu korumayan geçici bir alt kabukta yürütülmesidir. Çalışma dizinini daha kalıcı biçimde değiştirmek için %cd magic komutunu kullanabilirsiniz:


```python
# magic_cd.ipynb
In [14]: %cd ..
/home/jake/projects
```

*(Jupyter/IPython veya özel ortam gerekir — web sayfasında salt okunur.)*


Varsayılan olarak bunu % işareti olmadan bile kullanabilirsiniz:


```
In [15]: cd myproject
/home/jake/projects/myproject
```


Buna automagic (otomatik magic) denir; açık % olmadan bu tür komutları çalıştırma %automagic magic fonksiyonuyla açılıp kapatılabilir.

%cd dışında kullanılabilir kabuk benzeri magic fonksiyonlar: %cat, %cp, %env, %ls, %man, %mkdir, %more, %mv, %pwd, %rm ve %rmdir. automagic açıksa bunların hepsi % olmadan da kullanılabilir. Böylece IPython istemini neredeyse normal bir kabuk gibi kullanabilirsiniz:


```
In [16]: mkdir tmp

In [17]: ls
myproject.txt  tmp/

In [18]: cp myproject.txt tmp/

In [19]: ls tmp
myproject.txt

In [20]: rm -r tmp
```


Python oturumunuzla aynı terminal penceresinden kabuğa erişmek, iş akışlarınızda Python ile kabuğu daha az bağlam değiştirerek birleştirmenizi sağlar.
