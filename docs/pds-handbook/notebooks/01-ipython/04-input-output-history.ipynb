Orijinal: Input and Output History

Daha önce IPython kabuğunun önceki komutlara yukarı/aşağı ok veya Ctrl-p/Ctrl-n ile erişmenizi sağladığını gördünüz. Ayrıca hem kabukta hem defterlerde IPython, önceki komutların çıktısını ve komutların string sürümlerini almanın birkaç yolunu sunar. Bunları burada inceleyeceğiz.

## IPython'un In ve Out Nesneleri

IPython'un kullandığı In [1]: / Out[1]: istem stiline alışıyor olmalısınız. Bunlar yalnızca süs değildir: oturumdaki önceki girdi ve çıktılara nasıl erişeceğinize ipucu verirler. Şöyle bir oturum başlatalım:


```
In [1]: import math

In [2]: math.sin(2)
Out[2]: 0.9092974268256817

In [3]: math.cos(2)
Out[3]: -0.4161468365471424
```


Yerleşik math paketini içe aktardık, sonra 2 sayısının sinüs ve kosinüsünü hesapladık. Bu girdi ve çıktılar kabukta In/Out etiketleriyle gösterilir; ama daha fazlası var — IPython aslında In ve Out adlı Python değişkenleri oluşturur ve bunları geçmişi yansıtacak şekilde otomatik günceller:


```
In [4]: In
Out[4]: ['', 'import math', 'math.sin(2)', 'math.cos(2)', 'In']

In [5]: Out
Out[5]:
{2: 0.9092974268256817,
 3: -0.4161468365471424,
 4: ['', 'import math', 'math.sin(2)', 'math.cos(2)', 'In', 'Out']}
```


In nesnesi bir listedir; komutları sırayla tutar (listenin ilk öğesi yer tutucudur, böylece In [1] ilk komuta referans verebilir):


```
In [6]: print(In[1])
import math
```


Out nesnesi liste değil, girdi numaralarını çıktılarına eşleyen bir sözlüktür (varsa):


```
In [7]: print(Out[2])
0.9092974268256817
```


Tüm işlemlerin çıktısı yoktur: örneğin import ve print ifadeleri çıktıyı etkilemez. İkincisi şaşırtıcı gelebilir; print None döndüren bir fonksiyondur — kısalık için None döndüren komutlar Out'a eklenmez.

Bu, geçmiş sonuçlarla etkileşmek istediğinizde yararlıdır. Örneğin daha önce hesaplanan sonuçlarla sin(2) ** 2 + cos(2) ** 2 toplamını kontrol edelim:


```
In [8]: Out[2] ** 2 + Out[3] ** 2
Out[8]: 1.0
```


Sonuç bilinen trigonometrik özdeşlikten beklediğimiz gibi 1.0'dır. Burada bu sonuçlara gerek olmayabilir; çok pahalı bir hesaplama yapıp sonucu bir değişkene atamayı unuttuğunuzda çok işe yarar.

## Alt Çizgi Kısayolları ve Önceki Çıktılar

Standart Python kabuğunda önceki çıktıya erişim için tek basit kısayol vardır: _ (tek alt çizgi) son çıktıyla güncellenir. IPython'da da böyledir:


```
In [9]: print(_)
1.0
```


IPython biraz daha ileri gider — çift alt çizgi sondan bir önceki, üçlü alt çizgi sondan ikinci önceki çıktıya erişir (çıktısı olmayan komutlar atlanır):


```
In [10]: print(__)
-0.4161468365471424

In [11]: print(___)
0.9092974268256817
```


IPython burada durur: üçten fazla alt çizgi saymak zorlaşır; o noktada satır numarasıyla referans vermek daha kolaydır.

Bir kısayol daha: Out[X] için kısaltma _X (tek alt çizgi + satır numarası):


```
In [12]: Out[2]
Out[12]: 0.9092974268256817

In [13]: _2
Out[13]: 0.9092974268256817
```


## Çıktıyı Bastırma

Bazen bir ifadenin çıktısını bastırmak isteyebilirsiniz (en yaygın kullanım ileride göreceğimiz çizim komutlarıdır). Ya da komutun sonucunu çıktı geçmişinde saklamak istemezsiniz — bellekte diğer referanslar kalkınca serbest bırakılsın diye. En kolay yol satır sonuna noktalı virgül eklemektir:


```
In [14]: math.sin(2) + math.cos(2);
```


Sonuç sessizce hesaplanır; ekranda gösterilmez ve Out sözlüğüne eklenmez:


```
In [15]: 14 in Out
Out[15]: False
```


## İlgili Magic Komutlar

Önceki girdilerin bir grubuna birden erişmek için %history magic komutu çok yararlıdır. İlk üç girdiyi yazdırmak:


```
In [16]: %history -n 1-3
   1: import math
   2: math.sin(2)
   3: math.cos(2)
```


Her zamanki gibi seçenekler için %history? yazabilirsiniz (? işlevi için 1.1 Yardım ve Dokümantasyon). Diğer yararlı magic'ler: komut geçmişinin bir bölümünü yeniden çalıştıran %rerun ve geçmişin bir kısmını dosyaya kaydeden %save.
