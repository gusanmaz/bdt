Orijinal: Data Indexing and Selection

Önceki bölümde Series, DataFrame ve Index nesnelerini tanıdık.

Bölüm 2'de NumPy dizilerinde değerlere erişme, atama ve değiştirme araçlarını ayrıntılı inceledik. Bunlar arasında indeksleme (arr[2, 1]), dilimleme (arr[:, 1:5]), maskeleme (arr[arr > 0]), fancy indexing (arr[0, [1, 5]]) ve bunların birleşimleri (arr[:, [1, 5]]) vardı. Burada Pandas Series ve DataFrame nesnelerinde benzer erişim ve değiştirme kalıplarına bakacağız. NumPy kalıplarını biliyorsanız Pandas'taki karşılıkları çok tanıdık gelecektir; yine de bilmeniz gereken birkaç incelik vardır.

Önce tek boyutlu Series ile başlayıp ardından iki boyutlu DataFrame'e geçeceğiz.

## Series'te Veri Seçimi

Önceki bölümde gördüğünüz gibi Series hem tek boyutlu NumPy dizisine hem de standart Python sözlüğüne benzer. Bu iki örtüşen benzetmeyi aklınızda tutarsanız bu yapılarda indeksleme ve seçim kalıplarını anlamanız kolaylaşır.

### Series, Sözlük Olarak

Sözlük gibi Series, anahtar koleksiyonundan değer koleksiyonuna eşleme sağlar:


````python
# hucere_4.py
import pandas as pd
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data

````

```
In [1]:
a    0.25
b    0.50
c    0.75
d    1.00
dtype: float64
```


````python
# hucere_5.py
data['b']

````

```
In [2]:
0.5
```


Anahtar/indeks ve değerleri incelemek için sözlük benzeri Python ifadeleri ve yöntemleri de kullanılabilir:


````python
# hucere_7.py
'a' in data

````

```
In [3]:
True
```


````python
# hucere_8.py
data.keys()

````

```
In [4]:
Index(['a', 'b', 'c', 'd'], dtype='object')
```


````python
# hucere_9.py
list(data.items())

````

```
In [5]:
[('a', 0.25), ('b', 0.5), ('c', 0.75), ('d', 1.0)]
```


Series nesneleri sözlük benzeri sözdizimiyle de değiştirilebilir. Sözlüğe yeni anahtar atayarak genişlettiğiniz gibi, yeni indeks değerine atayarak Series'i genişletebilirsiniz:


````python
# hucere_11.py
data['e'] = 1.25
data

````

```
In [6]:
a    0.25
b    0.50
c    0.75
d    1.00
e    1.25
dtype: float64
```


Nesnelerin bu kolay değiştirilebilirliği kullanışlı bir özelliktir: Pandas arka planda bellek düzeni ve veri kopyalama kararlarını verir; kullanıcı genelde bu konularla uğraşmak zorunda kalmaz.

### Series, Tek Boyutlu Dizi Olarak

Series, bu sözlük benzeri arayüzün üzerine NumPy dizilerindeki gibi dilimleme, maskeleme ve fancy indexing ile dizi tarzı seçim sunar. Örnekler:


````python
# slicing by explicit index
data['a':'c']

````

```
In [7]:
a    0.25
b    0.50
c    0.75
dtype: float64
```


````python
# slicing by implicit integer index
data[0:2]

````

```
In [8]:
a    0.25
b    0.50
dtype: float64
```


````python
# masking
data[(data > 0.3) & (data < 0.8)]

````

```
In [9]:
b    0.50
c    0.75
dtype: float64
```


````python
# fancy indexing
data[['a', 'e']]

````

```
In [10]:
a    0.25
e    1.25
dtype: float64
```


Bunların içinde en çok kafa karıştıran dilimleme olabilir. Açık indeksle dilimlerken (ör. data['a':'c']) son indeks dilime dahildir; örtük indeksle dilimlerken (ör. data[0:2]) son indeks dilime dahil değildir.

### İndeksleyiciler: loc ve iloc

Series'in açık tamsayı indeksi varsa data[1] gibi indeksleme açık indeksi, data[1:3] gibi dilimleme ise örtük Python tarzı indeksi kullanır:


````python
# hucere_21.py
data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data

````

```
In [11]:
1    a
3    b
5    c
dtype: object
```


````python
# explicit index when indexing
data[1]

````

```
In [12]:
'a'
```


````python
# implicit index when slicing
data[1:3]

````

```
In [13]:
3    b
5    c
dtype: object
```


Tamsayı indekslerde bu karışıklık riski nedeniyle Pandas belirli indeksleme şemalarını açıkça sunan özel indeksleyici öznitelikleri sağlar. Bunlar fonksiyon değil; Series içindeki veriye belirli bir dilim arayüzü sunan özniteliklerdir.

loc özniteliği her zaman açık indeksi referans alan indeksleme ve dilimlemeye izin verir:


````python
# hucere_25.py
data.loc[1]

````

```
In [14]:
'a'
```


````python
# hucere_26.py
data.loc[1:3]

````

```
In [15]:
1    a
3    b
dtype: object
```


iloc özniteliği her zaman örtük Python tarzı indeksi referans alır:


````python
# hucere_28.py
data.iloc[1]

````

```
In [16]:
'b'
```


````python
# hucere_29.py
data.iloc[1:3]

````

```
In [17]:
3    b
5    c
dtype: object
```


Python'da “açık, örtükten iyidir” ilkesi geçerlidir. loc ve iloc'un açık doğası kodu okunaklı tutar; özellikle tamsayı indekslerde tutarlı kullanım, karışık indeksleme/dilimleme kurallarından kaynaklanan ince hataları önler.

## DataFrame'de Veri Seçimi

DataFrame hem iki boyutlu veya yapılandırılmış diziye hem de aynı indeksi paylaşan Series sözlüğüne benzer. Bu benzetmeler veri seçimini keşfederken yardımcı olur.

### DataFrame, Sözlük Olarak

İlk benzetme: DataFrame, ilişkili Series nesnelerinin sözlüğü. Eyalet alan ve nüfus örneğimize dönelim:


````python
# hucere_33.py
area = pd.Series({'California': 423967, 'Texas': 695662,
                  'Florida': 170312, 'New York': 141297,
                  'Pennsylvania': 119280})
pop = pd.Series({'California': 39538223, 'Texas': 29145505,
                 'Florida': 21538187, 'New York': 20201249,
                 'Pennsylvania': 13002700})
data = pd.DataFrame({'area':area, 'pop':pop})
data

````

```
In [18]:
                area       pop
California    423967  39538223
Texas         695662  29145505
Florida       170312  21538187
New York      141297  20201249
Pennsylvania  119280  13002700
```


DataFrame sütunlarını oluşturan Series'lere sütun adıyla sözlük tarzı erişilir:


````python
# hucere_35.py
data['area']

````

```
In [19]:
California      423967
Texas           695662
Florida         170312
New York        141297
Pennsylvania    119280
Name: area, dtype: int64
```


Eşdeğer olarak, string sütun adları için öznitelik tarzı erişim de mümkündür:


````python
# hucere_37.py
data.area

````

```
In [20]:
California      423967
Texas           695662
Florida         170312
New York        141297
Pennsylvania    119280
Name: area, dtype: int64
```


Bu kısayol kullanışlı olsa da her durumda çalışmaz! Sütun adları string değilse veya DataFrame yöntemleriyle çakışıyorsa öznitelik erişimi mümkün olmayabilir. Örneğin DataFrame'in bir pop yöntemi vardır; data.pop sütun yerine bu yönteme işaret eder:


````python
# hucere_39.py
data.pop is data["pop"]

````

```
In [21]:
False
```


Özellikle sütun atamayı öznitelikle yapmaktan kaçının (data.pop = z yerine data['pop'] = z kullanın).

Daha önceki Series gibi bu sözlük sözdizimi nesneyi değiştirmek için kullanılır; burada yeni sütun eklenir:


````python
# hucere_41.py
data['density'] = data['pop'] / data['area']
data

````

```
In [22]:
                area       pop     density
California    423967  39538223   93.257784
Texas         695662  29145505   41.896072
Florida       170312  21538187  126.463121
New York      141297  20201249  142.970120
Pennsylvania  119280  13002700  109.009893
```


Bu, Series nesneleri arasında eleman-eleman aritmetiğin sade sözdizimine bir ön izlemedir; ayrıntı için Pandas'ta İşlemler bölümüne bakın.

### DataFrame, İki Boyutlu Dizi Olarak

Daha önce belirtildiği gibi DataFrame geliştirilmiş iki boyutlu dizi olarak da görülebilir. Ham veri dizisine values özniteliğiyle bakılabilir:


````python
# hucere_44.py
data.values

````

```

       [6.95662000e+05, 2.91455050e+07, 4.18960717e+01],
       [1.70312000e+05, 2.15381870e+07, 1.26463121e+02],
       [1.41297000e+05, 2.02012490e+07, 1.42970120e+02],
       [1.19280000e+05, 1.30027000e+07, 1.09009893e+02]])
```


Bu bakışla DataFrame üzerinde birçok tanıdık dizi benzeri işlem yapılabilir; örneğin tam DataFrame transpoze edilerek satır ve sütunlar yer değiştirilebilir:


````python
# hucere_46.py
data.T

````

```
In [24]:

density  9.325778e+01  4.189607e+01  1.264631e+02  1.429701e+02  1.090099e+02
```


DataFrame indekslemesinde ise sütunların sözlük tarzı indekslenmesi onu doğrudan NumPy dizisi gibi ele almayı engeller. Özellikle diziye tek indeks verildiğinde satıra erişilir:


````python
# hucere_48.py
data.values[0]

````

```
In [25]:
array([4.23967000e+05, 3.95382230e+07, 9.32577842e+01])
```


DataFrame'e tek “indeks” verildiğinde ise sütuna erişilir:


````python
# hucere_50.py
data['area']

````

```
In [26]:
California      423967
Texas           695662
Florida         170312
New York        141297
Pennsylvania    119280
Name: area, dtype: int64
```


Dizi tarzı indeksleme için başka bir kural gerekir. Pandas yine loc ve iloc indeksleyicilerini kullanır. iloc ile alttaki dizi NumPy dizisi gibi indekslenir (örtük Python indeksi); sonuçta DataFrame indeks ve sütun etiketleri korunur:


````python
# hucere_52.py
data.iloc[:3, :2]

````

```
In [27]:
              area       pop
California  423967  39538223
Texas       695662  29145505
Florida     170312  21538187
```


loc ile veri açık indeks ve sütun adlarıyla dizi benzeri indekslenir:


````python
# hucere_54.py
data.loc[:'Florida', :'pop']

````

```
In [28]:
              area       pop
California  423967  39538223
Texas       695662  29145505
Florida     170312  21538187
```


Bu indeksleyiciler içinde tanıdık NumPy tarzı erişim kalıpları kullanılabilir; loc içinde maskeleme ve fancy indexing birleştirilebilir:


````python
# hucere_56.py
data.loc[data.density > 120, ['pop', 'density']]

````

```
In [29]:
               pop     density
Florida   21538187  126.463121
New York  20201249  142.970120
```


Bu indeksleme kuralları değer atamak veya değiştirmek için de kullanılır; NumPy'de alışık olduğunuz standart yolla:


````python
# hucere_58.py
data.iloc[0, 2] = 90
data

````

```
In [30]:
                area       pop     density
California    423967  39538223   90.000000
Texas         695662  29145505   41.896072
Florida       170312  21538187  126.463121
New York      141297  20201249  142.970120
Pennsylvania  119280  13002700  109.009893
```


Pandas veri manipülasyonunda akıcılık için basit bir DataFrame ile bu indeksleme yaklaşımlarının izin verdiği indeksleme, dilimleme, maskeleme ve fancy indexing türlerini denemenizi öneririm.

### Ek İndeksleme Kuralları

Önceki tartışmayla çelişiyor gibi görünen ama pratikte yararlı birkaç ek kural vardır. Önce: indeksleme sütunlara, dilimleme satırlara referans verir:


````python
# hucere_61.py
data['Florida':'New York']

````

```
In [31]:
            area       pop     density
Florida   170312  21538187  126.463121
New York  141297  20201249  142.970120
```


Bu dilimler indeks yerine satır numarasıyla da yapılabilir:


````python
# hucere_63.py
data[1:3]

````

```
In [32]:
           area       pop     density
Texas    695662  29145505   41.896072
Florida  170312  21538187  126.463121
```


Benzer şekilde doğrudan maskeleme sütun yerine satır bazında yorumlanır:


````python
# hucere_65.py
data[data.density > 120]

````

```
In [33]:
            area       pop     density
Florida   170312  21538187  126.463121
New York  141297  20201249  142.970120
```


Bu iki kural NumPy dizisindekine sözdizimsel olarak benzer; Pandas kurallarına tam uymasa da pratik faydaları nedeniyle dahil edilmiştir.
