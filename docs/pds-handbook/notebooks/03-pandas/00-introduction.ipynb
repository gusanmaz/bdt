Bölüm 2'de NumPy ve yoğun tipli dizileri verimli saklayıp işleyen ndarray nesnesini ayrıntılı ele aldık. Burada Pandas kütüphanesinin sunduğu veri yapılarına derinlemesine bakacağız. Pandas, NumPy üzerine kurulu daha yeni bir pakettir ve DataFrame için verimli bir uygulama sağlar.

DataFrame'ler, satır ve sütun etiketleri eklenmiş çok boyutlu diziler gibidir; çoğu zaman heterojen tipler ve/veya eksik veri içerir. Etiketli veri için uygun bir depolama arayüzü sunmanın yanı sıra Pandas, veritabanı çerçevelerinde ve hesap tablolarında tanıdık güçlü veri işlemlerini uygular.

Gördüğümüz gibi NumPy'nin ndarray yapısı, sayısal hesaplamada görülen düzenli veri türleri için temel özellikler sunar. Bu amaca iyi hizmet eder; ancak daha fazla esneklik gerektiğinde (veriye etiket ekleme, eksik veriyle çalışma vb.) ve eleman bazlı broadcasting'e iyi uymayan işlemlerde (gruplama, pivot vb.) sınırları belirginleşir — bunların her biri dünyadaki daha az yapılandırılmış veriyi analiz etmenin önemli parçalarıdır. Pandas, özellikle Series ve DataFrame nesneleriyle NumPy dizi yapısının üzerine inşa edilir ve bir veri bilimcinin zamanının büyük kısmını alan bu tür data munging görevlerine verimli erişim sağlar.
