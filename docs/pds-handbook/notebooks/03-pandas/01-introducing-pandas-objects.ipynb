## Pandas Series Nesnesi

Pandas Series, indekslenmiş veriden oluşan tek boyutlu bir dizidir. Liste veya dizi ile şöyle oluşturulabilir:


````python
# hucere_4.py
data = pd.Series([0.25, 0.5, 0.75, 1.0])
data

````

```
In [2]:
0    0.25
1    0.50
2    0.75
3    1.00
dtype: float64
```


Series, bir değer dizisini açık bir indeks dizisiyle birleştirir; values ve index öznitelikleriyle erişilir. values tanıdık bir NumPy dizisidir:


````python
# hucere_6.py
data.values

````

```
In [3]:
array([0.25, 0.5 , 0.75, 1.  ])
```


index, pd.Index tipinde dizi benzeri bir nesnedir; birazdan ayrıntılandıracağız:


````python
# hucere_8.py
data.index

````

```
In [4]:
RangeIndex(start=0, stop=4, step=1)
```


NumPy dizisinde olduğu gibi, veriye ilişkili indeks üzerinden Python köşeli parantez gösterimiyle erişilebilir:


````python
# hucere_10.py
data[1]

````

```
In [5]:
0.5
```


````python
# hucere_11.py
data[1:3]

````

```
In [6]:
1    0.50
2    0.75
dtype: float64
```


Göreceğimiz üzere Pandas Series, taklit ettiği tek boyutlu NumPy dizisinden çok daha genel ve esnektir.

### Series, Genelleştirilmiş NumPy Dizisi

Şimdiye kadar gördüklerimizden Series nesnesi tek boyutlu bir NumPy dizisiyle neredeyse değiştirilebilir görünebilir. Temel fark şudur: NumPy dizisinde değerlere erişmek için örtük tamsayı indeks kullanılır; Pandas Series'te ise değerlerle birlikte açık bir indeks tanımlıdır.

Bu açık indeks, Series'e ek yetenekler verir. Örneğin indeks tamsayı olmak zorunda değildir; istenen herhangi bir türde olabilir. İstersek string indeks kullanabiliriz:


````python
# hucere_15.py
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data

````

```
In [7]:
a    0.25
b    0.50
c    0.75
d    1.00
dtype: float64
```


Öğe erişimi beklendiği gibi çalışır:


````python
# hucere_17.py
data['b']

````

```
In [8]:
0.5
```


Ardışık olmayan veya sıralı olmayan indeksler de kullanılabilir:


````python
# hucere_19.py
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=[2, 5, 3, 7])
data

````

```
In [9]:
2    0.25
5    0.50
3    0.75
7    1.00
dtype: float64
```


````python
# hucere_20.py
data[5]

````

```
In [10]:
0.5
```


### Series, Özelleştirilmiş Sözlük

Bu bakışla Pandas Series, Python sözlüğünün bir özelleşmesi gibi düşünülebilir. Sözlük key → değer eşlemesi yapar; Series ise tipli anahtarları tipli değerlere eşler. Bu tip bilgisi önemlidir: NumPy dizisinin arkasındaki tip odaklı derlenmiş kod, belirli işlemlerde Python listesinden daha verimli kılar; Series'in tip bilgisi de benzer şekilde Python sözlüklerinden daha verimli olmasını sağlar.

Series-sözlük benzetmesi, doğrudan Python sözlüğünden Series oluşturarak netleşir — burada 2020 nüfus sayımına göre ABD'nin en kalabalık beş eyaleti:


````python
# hucere_22.py
population_dict = {'California': 39538223, 'Texas': 29145505,
                   'Florida': 21538187, 'New York': 20201249,
                   'Pennsylvania': 13002700}
population = pd.Series(population_dict)
population

````

```
In [11]:
California      39538223
Texas           29145505
Florida         21538187
New York        20201249
Pennsylvania    13002700
dtype: int64
```


Buradan tipik sözlük tarzı öğe erişimi yapılabilir:


````python
# hucere_24.py
population['California']

````

```
In [12]:
39538223
```


Sözlükten farklı olarak Series dilimleme gibi dizi tarzı işlemleri de destekler:


````python
# hucere_26.py
population['California':'Florida']

````

```
In [13]:
California    39538223
Texas         29145505
Florida       21538187
dtype: int64
```


Pandas indeksleme ve dilimlemenin bazı inceliklerini Veri İndeksleme ve Seçimi bölümünde ele alacağız.

### Series Nesneleri Oluşturma

Sıfırdan Series oluşturmanın birkaç yolunu gördük; hepsi şu kalıba benzer:

pd.Series(data, index=index)

index isteğe bağlıdır; data birçok türden biri olabilir.

Örneğin data liste veya NumPy dizisi olabilir; bu durumda index varsayılan olarak tamsayı dizisidir:


````python
# hucere_29.py
pd.Series([2, 4, 6])

````

```
In [14]:
0    2
1    4
2    6
dtype: int64
```


Ya da data bir skaler olabilir; belirtilen indekste tekrarlanır:


````python
# hucere_31.py
pd.Series(5, index=[100, 200, 300])

````

```
In [15]:
100    5
200    5
300    5
dtype: int64
```


Ya da sözlük olabilir; index varsayılan olarak sözlük anahtarlarıdır:


````python
# hucere_33.py
pd.Series({2:'a', 1:'b', 3:'c'})

````

```
In [16]:
2    a
1    b
3    c
dtype: object
```


Her durumda indeks, sırayı veya kullanılacak anahtar alt kümesini açıkça belirlemek için ayarlanabilir:


````python
# hucere_35.py
pd.Series({2:'a', 1:'b', 3:'c'}, index=[1, 2])

````

```
In [17]:
1    b
2    a
dtype: object
```


## Pandas DataFrame Nesnesi

Pandas'taki bir sonraki temel yapı DataFrame'dir. Önceki bölümdeki Series gibi DataFrame hem NumPy dizisinin genellemesi hem de Python sözlüğünün özelleşmesi olarak düşünülebilir. Her iki bakış açısını da inceleyeceğiz.

### DataFrame, Genelleştirilmiş NumPy Dizisi

Series açık indeksli tek boyutlu dizi analoguysa, DataFrame açık satır ve sütun indeksli iki boyutlu dizi analogudur. İki boyutlu diziyi hizalı tek boyutlu sütunların sıralı dizisi gibi düşünürseniz, DataFrame'i hizalı Series nesnelerinin dizisi olarak düşünebilirsiniz. “Hizalı” derken aynı indeksi paylaştıklarını kastediyoruz.

Bunu göstermek için önceki bölümdeki beş eyaletin alanını (km²) listeleyen yeni bir Series oluşturalım:


````python
# hucere_38.py
area_dict = {'California': 423967, 'Texas': 695662, 'Florida': 170312, 
             'New York': 141297, 'Pennsylvania': 119280}
area = pd.Series(area_dict)
area

````

```
In [18]:
California      423967
Texas           695662
Florida         170312
New York        141297
Pennsylvania    119280
dtype: int64
```


Bunu ve önceki population Series'ini kullanarak bu bilgiyi içeren tek bir iki boyutlu nesne oluşturabiliriz:


````python
# hucere_40.py
states = pd.DataFrame({'population': population,
                       'area': area})
states

````

```
In [19]:
              population    area
California      39538223  423967
Texas           29145505  695662
Florida         21538187  170312
New York        20201249  141297
Pennsylvania    13002700  119280
```


Series gibi DataFrame'in de indeks etiketlerine erişen bir index özniteliği vardır:


````python
# hucere_42.py
states.index

````

```
In [20]:
Index(['California', 'Texas', 'Florida', 'New York', 'Pennsylvania'], dtype='object')
```


Ayrıca sütun etiketlerini tutan Index tipinde bir columns özniteliği vardır:


````python
# hucere_44.py
states.columns

````

```
In [21]:
Index(['population', 'area'], dtype='object')
```


Böylece DataFrame, hem satır hem sütunların veriye erişim için genelleştirilmiş indekse sahip olduğu iki boyutlu NumPy dizisinin genellemesi olarak görülebilir.

### DataFrame, Özelleştirilmiş Sözlük

Benzer şekilde DataFrame bir sözlük özelleşmesi olarak da düşünülebilir. Sözlük anahtar → değer eşler; DataFrame sütun adı → sütun verisinin Series'i eşler. Örneğin 'area' istendiğinde daha önce gördüğümüz alan Series'i döner:


````python
# hucere_47.py
states['area']

````

```
In [22]:
California      423967
Texas           695662
Florida         170312
New York        141297
Pennsylvania    119280
Name: area, dtype: int64
```


Burada kafa karıştırıcı bir nokta olabilir: iki boyutlu NumPy dizisinde data[0] ilk satırı döndürür; DataFrame'de data['col0'] ilk sütunu döndürür. Bu yüzden DataFrame'i genelleştirilmiş dizi yerine genelleştirilmiş sözlük olarak düşünmek genelde daha iyidir; her iki bakış da yararlı olabilir. DataFrame için daha esnek indekslemeyi Veri İndeksleme ve Seçimi bölümünde inceleyeceğiz.

### DataFrame Nesneleri Oluşturma

Pandas DataFrame birçok yolla oluşturulabilir. Burada birkaç örneği ele alıyoruz.

#### Tek bir Series nesnesinden

DataFrame, Series nesnelerinin koleksiyonudur; tek sütunlu DataFrame tek bir Series'ten oluşturulabilir:


````python
# hucere_51.py
pd.DataFrame(population, columns=['population'])

````

```
In [23]:
              population
California      39538223
Texas           29145505
Florida         21538187
New York        20201249
Pennsylvania    13002700
```


#### Sözlük listesinden

Herhangi bir sözlük listesi DataFrame yapılabilir. Basit bir list comprehension ile veri üretelim:


````python
# hucere_53.py
data = [{'a': i, 'b': 2 * i}
        for i in range(3)]
pd.DataFrame(data)

````

```
In [24]:
   a  b
0  0  0
1  1  2
2  2  4
```


Sözlüklerde bazı anahtarlar eksik olsa bile Pandas eksikleri NaN (Not a Number) ile doldurur; bkz. Eksik Veri:


````python
# hucere_55.py
pd.DataFrame([{'a': 1, 'b': 2}, {'b': 3, 'c': 4}])

````

```
In [25]:
     a  b    c
0  1.0  2  NaN
1  NaN  3  4.0
```


Daha önce gördüğümüz gibi DataFrame, Series sözlüğünden de oluşturulabilir:


````python
# hucere_57.py
pd.DataFrame({'population': population,
              'area': area})

````

```
In [26]:
              population    area
California      39538223  423967
Texas           29145505  695662
Florida         21538187  170312
New York        20201249  141297
Pennsylvania    13002700  119280
```


#### İki boyutlu NumPy dizisinden

İki boyutlu veri dizisi verildiğinde istenen sütun ve indeks adlarıyla DataFrame oluşturulabilir. Belirtilmezse her biri için tamsayı indeks kullanılır:


````python
# hucere_59.py
pd.DataFrame(np.random.rand(3, 2),
             columns=['foo', 'bar'],
             index=['a', 'b', 'c'])

````

```
In [27]:
        foo       bar
a  0.471098  0.317396
b  0.614766  0.305971
c  0.533596  0.512377
```


#### NumPy yapılandırılmış dizisinden

Yapılandırılmış dizileri Yapılandırılmış Veri: NumPy Yapılandırılmış Dizileri bölümünde ele aldık. Pandas DataFrame yapılandırılmış diziye çok benzer çalışır ve doğrudan ondan oluşturulabilir:


````python
# hucere_61.py
A = np.zeros(3, dtype=[('A', 'i8'), ('B', 'f8')])
A

````

```
In [28]:
array([(0, 0.), (0, 0.), (0, 0.)], dtype=[('A', '<i8'), ('B', '<f8')])
```

````python
# hucere_62.py
pd.DataFrame(A)

````

```
   A    B

2  0  0.0
```


## Pandas Index Nesnesi

Gördüğünüz gibi Series ve DataFrame veriye referans ve değişiklik için açık bir indeks içerir. Index kendi başına ilginç bir yapıdır; değiştirilemez dizi veya sıralı küme (teknik olarak çok küme — Index tekrarlı değer içerebilir) olarak düşünülebilir. Bu görünümler Index üzerindeki işlemlere ilginç sonuçlar verir. Basit örnek: tamsayı listesinden Index oluşturalım:


````python
# hucere_64.py
ind = pd.Index([2, 3, 5, 7, 11])
ind

````

```
In [30]:
Int64Index([2, 3, 5, 7, 11], dtype='int64')
```


### Index, Değiştirilemez Dizi

Index birçok yönden dizi gibi davranır. Standart Python indeks gösterimiyle değer veya dilim alınabilir:


````python
# hucere_66.py
ind[1]

````

```
In [31]:
3
```


````python
# hucere_67.py
ind[::2]

````

```
In [32]:
Int64Index([2, 5, 11], dtype='int64')
```


Index nesneleri NumPy dizilerinden tanıdık birçok özniteliğe de sahiptir:


````python
# hucere_69.py
print(ind.size, ind.shape, ind.ndim, ind.dtype)

````

```
In [33]:
5 (5,) 1 int64
```


Index ile NumPy dizisi arasındaki farklardan biri indekslerin değiştirilemez olmasıdır — normal yollarla değiştirilemezler:


````python
# hucere_71.py
ind[1] = 0

````

```
In [34]:
TypeError: Index does not support mutable operations
```


Bu değişmezlik, birden fazla DataFrame ve dizi arasında indeks paylaşmayı daha güvenli kılar; istemeden indeks değişikliğinin yan etkileri olmaz.

### Index, Sıralı Küme

Pandas nesneleri veri kümeleri arasında birleştirme gibi küme aritmetiğine dayanan işlemleri kolaylaştıracak şekilde tasarlanmıştır. Index, Python'un yerleşik set yapısının birçok kuralını izler; birleşim, kesişim, fark ve diğer kombinasyonlar tanıdık biçimde hesaplanabilir:


````python
# hucere_74.py
indA = pd.Index([1, 3, 5, 7, 9])
indB = pd.Index([2, 3, 5, 7, 11])

````

````python
# hucere_75.py
indA.intersection(indB)

````

```
In [36]:
Int64Index([3, 5, 7], dtype='int64')
```


````python
# hucere_76.py
indA.union(indB)

````

```
In [37]:
Int64Index([1, 2, 3, 5, 7, 9, 11], dtype='int64')
```


````python
# hucere_77.py
indA.symmetric_difference(indB)

````

```
In [38]:
Int64Index([1, 2, 9, 11], dtype='int64')
```
