## NumPy dizi öznitelikleri

Önce yararlı dizi özniteliklerini konuşalım. Bir, iki ve üç boyutlu rastgele diziler tanımlayarak başlayacağız. Her çalıştırmada aynı rastgele dizilerin üretilmesi için NumPy'nin rastgele sayı üretecini sabit bir tohum (seed) ile başlatacağız:


````python
# dizi_tanimlama.py
import numpy as np
rng = np.random.default_rng(seed=1701)  # tekrarlanabilirlik için tohum

x1 = rng.integers(10, size=6)           # tek boyutlu dizi
x2 = rng.integers(10, size=(3, 4))      # iki boyutlu dizi
x3 = rng.integers(10, size=(3, 4, 5))   # üç boyutlu dizi

````

Her dizinin ndim (boyut sayısı), shape (her boyutun uzunluğu), size (toplam eleman sayısı) ve dtype (eleman tipi) gibi öznitelikleri vardır:


````python
# dizi_oznitelikleri.py
print("x3 ndim: ", x3.ndim)
print("x3 shape:", x3.shape)
print("x3 size: ", x3.size)
print("dtype:   ", x3.dtype)

````

Veri tipleri hakkında daha fazla bilgi için 2.1 Python'da Veri Tiplerini Anlamak bölümüne bakın.

## Dizi indeksleme: Tek elemana erişim

Python'un standart liste indekslemesine aşina iseniz NumPy indekslemesi tanıdık gelecektir. Tek boyutlu bir dizide, sıfırdan sayarak i. değere köşeli parantez içinde istenen indeks yazılarak erişilir — tıpkı Python listelerinde olduğu gibi:


````python
# indeks_1b.py
print(x1)
print(x1[0])
print(x1[4])

````

Dizinin sonundan indekslemek için negatif indeksler kullanılabilir:


````python
# indeks_negatif.py
print(x1[-1])
print(x1[-2])

````

Çok boyutlu bir dizide elemanlara virgülle ayrılmış (satır, sütun) demeti ile erişilir:


````python
# indeks_2b.py
print(x2)
print(x2[0, 0])
print(x2[2, 0])
print(x2[2, -1])

````

Değerler yukarıdaki indeks gösterimleriyle de değiştirilebilir:


````python
# indeks_yazma.py
x2[0, 0] = 12
print(x2)

````

Python listelerinin aksine NumPy dizilerinin sabit bir tipi vardır. Örneğin tamsayı dizisine kayan noktalı değer yazmaya çalışırsanız değer sessizce kesilir (truncate). Bu davranışa hazırlıksız yakalanmayın!


````python
# tip_kesme.py
x1[0] = 3.14159  # bu kesilecek!
print(x1)

````

## Dizi dilimleme: Alt dizilere erişim

Tekil dizi elemanlarına köşeli parantezle erişebildiğimiz gibi, iki nokta (:) karakteriyle işaretlenen dilim gösterimiyle alt dizilere de erişebiliriz. NumPy dilim sözdizimi standart Python listesini izler; x dizisinin bir dilimine erişmek için:


```python
# slice_syntax.py
x[start:stop:step]
```

*(Jupyter/IPython veya özel ortam gerekir — web sayfasında salt okunur.)*


Bunlardan herhangi biri belirtilmezse varsayılanlar start=0, stop=<boyut>, step=1 olur. Tek ve çok boyutta alt dizilere erişimin birkaç örneğine bakalım.

### Tek boyutlu alt diziler

Tek boyutlu alt dizilerde eleman erişimine örnekler:


````python
# slice_1b.py
print(x1)
print(x1[:3])      # ilk üç eleman
print(x1[3:])      # 3. indeksten sonraki elemanlar
print(x1[1:4])      # orta alt dizi
print(x1[::2])      # her ikinci eleman
print(x1[1::2])     # 1. indeksten başlayarak her ikinci eleman

````

step değerinin negatif olduğu durum kafa karıştırıcı olabilir. Bu durumda start ve stop varsayılanları yer değiştirir. Bu, diziyi ters çevirmenin pratik bir yoludur:


````python
# slice_ters.py
print(x1[::-1])     # tüm elemanlar, ters sırada
print(x1[4::-2])    # 4. indeksten başlayarak her ikinci eleman, ters

````

### Çok boyutlu alt diziler

Çok boyutlu dilimler aynı şekilde çalışır; birden fazla dilim virgülle ayrılır. Örneğin:


````python
# slice_2b.py
print(x2)
print(x2[:2, :3])   # ilk iki satır ve üç sütun
print(x2[:3, ::2])   # üç satır, her ikinci sütun
print(x2[::-1, ::-1])  # tüm satır ve sütunlar, ters

````

#### Dizi satır ve sütunlarına erişim

Sık ihtiyaç duyulan bir işlem, dizinin tek bir satırına veya sütununa erişmektir. Bu, indeksleme ve dilimlemeyi birleştirerek, tek iki nokta (:) ile işaretlenen boş dilim kullanılarak yapılır:


````python
# satir_sutun.py
print(x2[:, 0])   # x2'nin ilk sütunu
print(x2[0, :])   # x2'nin ilk satırı

````

Satır erişiminde boş dilim daha kısa sözdizimi için atlanabilir:


````python
# satir_kisa.py
print(x2[0])      # x2[0, :] ile eşdeğer

````

### Alt diziler kopya değil görünümdür

Python liste dilimlerinin aksine NumPy dizi dilimleri dizinin kopyası değil görünümü (view) olarak döner. Önceki iki boyutlu dizimize bakalım:


````python
# view_ornek_1.py
print(x2)

````

Bundan 2×2'lik bir alt dizi çıkaralım:


````python
# view_ornek_2.py
x2_sub = x2[:2, :2]
print(x2_sub)

````

Bu alt diziyi değiştirirsek orijinal dizinin de değiştiğini görürüz:


````python
# view_ornek_3.py
x2_sub[0, 0] = 99
print(x2_sub)
print(x2)

````

Bazı kullanıcılar bunu şaşırtıcı bulabilir; ancak büyük veri kümelerinde alttaki veri tamponunu kopyalamadan parçalara erişip işlemek avantajlıdır.

### Dizi kopyaları oluşturma

Dizi görünümlerinin güzel özelliklerine rağmen bazen dizinin veya alt dizinin verisini açıkça kopyalamak yararlıdır. En kolay yol copy yöntemidir:


````python
# copy_ornek_1.py
x2_sub_copy = x2[:2, :2].copy()
print(x2_sub_copy)

````

Bu alt diziyi değiştirirsek orijinal dizi etkilenmez:


````python
# copy_ornek_2.py
x2_sub_copy[0, 0] = 42
print(x2_sub_copy)
print(x2)

````

## Dizi yeniden şekillendirme

Bir diğer yararlı işlem, reshape yöntemiyle dizileri yeniden şekillendirmektir. Örneğin 1'den 9'a kadar sayıları 3×3 bir ızgaraya yerleştirmek için:


````python
# reshape_ornek.py
grid = np.arange(1, 10).reshape(3, 3)
print(grid)

````

Bunun çalışması için başlangıç dizisinin boyutu yeniden şekillendirilmiş dizinin boyutuyla eşleşmelidir; çoğu durumda reshape başlangıç dizisinin kopyasız bir görünümünü döndürür.

Yaygın bir yeniden şekillendirme işlemi, tek boyutlu diziyi iki boyutlu satır veya sütun vektörüne dönüştürmektir:


````python
# reshape_vektor.py
x = np.array([1, 2, 3])
print(x.reshape((1, 3)))   # reshape ile satır vektörü
print(x.reshape((3, 1)))   # reshape ile sütun vektörü

````

Bunun pratik kısayolu dilim sözdiziminde np.newaxis kullanmaktır:


````python
# newaxis_ornek.py
print(x[np.newaxis, :])   # newaxis ile satır vektörü
print(x[:, np.newaxis])   # newaxis ile sütun vektörü

````

Bu kalıbı kitabın geri kalanında sık kullanacağız.

## Dizi birleştirme ve bölme

Önceki tüm rutinler tek diziler üzerinde çalıştı. NumPy ayrıca birden fazla diziyi birleştirmek ve tek bir diziyi parçalara ayırmak için araçlar da sunar.

### Dizi birleştirme

NumPy'de iki dizinin birleştirilmesi (concatenation) öncelikle np.concatenate, np.vstack ve np.hstack rutinleriyle yapılır. np.concatenate ilk argüman olarak demet veya dizi listesi alır:


````python
# concat_1d.py
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])
print(np.concatenate([x, y]))

````

İkiden fazla diziyi de aynı anda birleştirebilirsiniz:


````python
# concat_uc.py
z = np.array([99, 99, 99])
print(np.concatenate([x, y, z]))

````

İki boyutlu diziler için de kullanılabilir:


````python
# concat_2d.py
grid = np.array([[1, 2, 3],
                 [4, 5, 6]])

# birinci eksen boyunca birleştir
print(np.concatenate([grid, grid]))

# ikinci eksen boyunca birleştir (sıfır indeksli)
print(np.concatenate([grid, grid], axis=1))

````

Karışık boyutlu dizilerle çalışırken np.vstack (dikey yığma) ve np.hstack (yatay yığma) daha anlaşılır olabilir:


````python
# dizileri dikey yığ
print(np.vstack([x, grid]))

# dizileri yatay yığ
y2 = np.array([[99],
              [99]])
print(np.hstack([grid, y2]))

````

Benzer şekilde daha yüksek boyutlu diziler için np.dstack dizileri üçüncü eksen boyunca yığar.

### Dizi bölme

Birleştirmenin tersi bölmedir; np.split, np.hsplit ve np.vsplit fonksiyonlarıyla uygulanır. Her biri için bölme noktalarını veren indeks listesi geçilebilir:


````python
# split_1d.py
x = [1, 2, 3, 99, 99, 3, 2, 1]
x1, x2, x3 = np.split(x, [3, 5])
print(x1, x2, x3)

````

N bölme noktası N + 1 alt dizi üretir. İlgili np.hsplit ve np.vsplit fonksiyonları benzerdir:


````python
# split_2d.py
grid = np.arange(16).reshape((4, 4))
print(grid)

upper, lower = np.vsplit(grid, [2])
print(upper)
print(lower)

left, right = np.hsplit(grid, [2])
print(left)
print(right)

````

Benzer şekilde daha yüksek boyutlu diziler için np.dsplit dizileri üçüncü eksen boyunca böler.
